## 5B – Core Utility Functions

These helper functions are reused throughout the application. They convert financial amounts between different time periods and calculate the user's share of shared expenses.

````python
def normalise_frequency(amount, frequency, budget_period="weekly"):
    """
    Convert an amount into the selected budgeting period.
    """

    if amount < 0:
        raise ValueError("Amount cannot be negative.")

    frequency = frequency.lower()
    budget_period = budget_period.lower()

    # Convert the original amount to a yearly amount first
    if frequency == "weekly":
        yearly_amount = amount * 52

    elif frequency == "fortnightly":
        yearly_amount = amount * 26

    elif frequency == "monthly":
        yearly_amount = amount * 12

    elif frequency == "yearly":
        yearly_amount = amount

    else:
        raise ValueError(
            "Frequency must be weekly, fortnightly, monthly or yearly."
        )

    # Convert yearly amount to the selected budget period
    if budget_period == "weekly":
        return yearly_amount / 52

    elif budget_period == "fortnightly":
        return yearly_amount / 26

    elif budget_period == "monthly":
        return yearly_amount / 12

    else:
        raise ValueError(
            "Budget period must be weekly, fortnightly or monthly."
        )


def calculate_user_share(total_amount, share_percent=100):
    """
    Calculate the user's share of a shared expense.
    """

    if total_amount < 0:
        raise ValueError("Expense amount cannot be negative.")

    if share_percent < 0 or share_percent > 100:
        raise ValueError(
            "Share percentage must be between 0 and 100."
        )

    return total_amount * share_percent / 100
````

## 5C – Income Functions

These functions calculate how much income is actually available to the user's budget. They support full, percentage-based, fixed and unavailable income, while keeping restricted income separate from general spending money.

````python
def calculate_available_income(
    amount,
    availability_type="all",
    availability_value=None
):
    """
    Calculate how much of an income source is available to the budget.
    """

    if amount < 0:
        raise ValueError("Income amount cannot be negative.")

    availability_type = availability_type.lower()

    if availability_type == "all":
        return amount

    elif availability_type == "percent":
        if availability_value is None:
            raise ValueError(
                "Please provide an availability percentage."
            )

        if availability_value < 0 or availability_value > 100:
            raise ValueError(
                "Percentage must be between 0 and 100."
            )

        return amount * availability_value / 100

    elif availability_type == "fixed":
        if availability_value is None:
            raise ValueError(
                "Please provide a fixed available amount."
            )

        if availability_value < 0 or availability_value > amount:
            raise ValueError(
                "Fixed available amount must be between 0 and the income amount."
            )

        return availability_value

    elif availability_type == "none":
        return 0

    else:
        raise ValueError(
            "Availability type must be all, percent, fixed or none."
        )


def calculate_income_totals(
    income_sources,
    budget_period="weekly"
):
    """
    Calculate general and restricted usable income.
    """

    general_income = 0
    restricted_income = 0

    for source in income_sources:

        available_amount = calculate_available_income(
            source["amount"],
            source.get("availability_type", "all"),
            source.get("availability_value")
        )

        normalised_amount = normalise_frequency(
            available_amount,
            source["frequency"],
            budget_period
        )

        if source.get("restricted", False):
            restricted_income += normalised_amount
        else:
            general_income += normalised_amount

    return general_income, restricted_income


def apply_restricted_income(
    expense_amount,
    restricted_amount
):
    """
    Apply restricted income to a related expense.
    """

    if expense_amount < 0 or restricted_amount < 0:
        raise ValueError("Amounts cannot be negative.")

    amount_used = min(
        expense_amount,
        restricted_amount
    )

    general_budget_amount = (
        expense_amount - amount_used
    )

    restricted_left = (
        restricted_amount - amount_used
    )

    return (
        amount_used,
        general_budget_amount,
        restricted_left
    )
````

## 5D – Expense Functions

These functions calculate the user's real share of expenses, convert them into a common budgeting period, and prepare expense data in a consistent format for later analysis.

````python
def clean_expense_data(expenses):
    """
    Standardise expense records so later functions can use them safely.
    """

    cleaned_expenses = []

    for expense in expenses:
        cleaned = expense.copy()

        if "sharing_percentage" in cleaned:
            cleaned["share_percent"] = cleaned.pop(
                "sharing_percentage"
            )

        cleaned.setdefault("share_percent", 100)
        cleaned.setdefault("status", "confirmed")
        cleaned.setdefault("classification", "essential flexible")
        cleaned.setdefault("pattern", "variable")

        cleaned_expenses.append(cleaned)

    return cleaned_expenses


def calculate_total_expenses(
    expenses,
    budget_period="weekly"
):
    """
    Calculate total confirmed expenses for the selected budget period.
    """

    total = 0

    for expense in expenses:

        if expense.get("status", "confirmed") != "confirmed":
            continue

        user_share = calculate_user_share(
            expense["amount"],
            expense.get("share_percent", 100)
        )

        normalised_amount = normalise_frequency(
            user_share,
            expense["frequency"],
            budget_period
        )

        total += normalised_amount

    return total


def calculate_money_remaining(
    general_income,
    total_expenses
):
    """
    Calculate money remaining after expenses.
    """

    return general_income - total_expenses


def calculate_suggested_buffer(
    general_income,
    percentage=5
):
    """
    Suggest a safety buffer based on available income.
    """

    if general_income < 0:
        raise ValueError(
            "Income cannot be negative."
        )

    if percentage < 0 or percentage > 100:
        raise ValueError(
            "Buffer percentage must be between 0 and 100."
        )

    return general_income * percentage / 100


def calculate_money_for_goals(
    money_remaining,
    buffer_amount
):
    """
    Calculate how much money is available for saving goals.
    """

    if buffer_amount < 0:
        raise ValueError(
            "Buffer amount cannot be negative."
        )

    return max(
        0,
        money_remaining - buffer_amount
    )
````

## 5E – Savings Goal Functions

These functions manage saving goals, calculate how much must be saved each week, check whether the plan is achievable, and update goal progress over time.

````python
def calculate_weeks_left(deadline, start_date=None):
    """
    Calculate how many weekly saving periods remain
    until the goal deadline.
    """

    if start_date is None:
        start_date = date.today()

    deadline_date = datetime.strptime(
        deadline,
        "%Y-%m-%d"
    ).date()

    days_left = (deadline_date - start_date).days

    if days_left < 0:
        return 0

    return math.ceil(days_left / 7)


def clean_goal_data(goals):
    """
    Prepare goal data for the calculation engine.
    """

    cleaned_goals = []

    for goal in goals:
        cleaned = goal.copy()

        if cleaned.get("current_saved") is None:
            cleaned["current_saved"] = 0

        cleaned["status"] = cleaned.get(
            "status",
            "active"
        )

        cleaned_goals.append(cleaned)

    return cleaned_goals


def update_goal_requirements(goals):
    """
    Calculate remaining money and required saving
    per week for each goal.
    """

    for goal in goals:

        remaining_amount = max(
            goal["target_amount"] - goal["current_saved"],
            0
        )

        goal["remaining_amount"] = remaining_amount

        if goal.get("status") != "active":
            goal["required_per_period"] = 0
            continue

        periods_left = goal.get("periods_left")

        if remaining_amount == 0:
            goal["required_per_period"] = 0
            goal["status"] = "completed"

        elif periods_left is None:
            goal["required_per_period"] = 0

        elif periods_left <= 0:
            goal["required_per_period"] = None

        else:
            goal["required_per_period"] = (
                remaining_amount / periods_left
            )

    return goals
````

````python
def check_updated_plan(
    goals,
    money_for_goals
):
    """
    Check whether the active saving goals are achievable.
    """

    total_required = 0
    overdue_goals = []

    for goal in goals:

        if goal.get("status") != "active":
            continue

        required = goal.get("required_per_period")

        if required is None:
            overdue_goals.append(goal["name"])

        else:
            total_required += required

    if overdue_goals:
        status = "needs attention"

    elif total_required <= money_for_goals:
        status = "achievable"

    else:
        status = "not achievable"

    surplus = max(
        0,
        money_for_goals - total_required
    )

    shortfall = max(
        0,
        total_required - money_for_goals
    )

    return {
        "status": status,
        "total_required": total_required,
        "surplus": surplus,
        "shortfall": shortfall,
        "overdue_goals": overdue_goals
    }


def allocate_goal_savings(
    goals,
    money_for_goals
):
    """
    Allocate the minimum weekly saving needed
    to each active goal.
    """

    allocations = {}

    active_goals = [
        goal for goal in goals
        if goal.get("status") == "active"
    ]

    total_required = 0

    for goal in active_goals:

        required = goal.get("required_per_period")

        if required is not None:
            total_required += required

    if total_required == 0:
        return {
            "allocations": allocations,
            "extra_money": money_for_goals,
            "shortfall": 0
        }

    if money_for_goals >= total_required:

        for goal in active_goals:
            required = goal.get("required_per_period")

            if required is not None:
                allocations[goal["name"]] = required

        return {
            "allocations": allocations,
            "extra_money": money_for_goals - total_required,
            "shortfall": 0
        }

    # If there is not enough money,
    # allocate proportionally.
    for goal in active_goals:

        required = goal.get("required_per_period")

        if required is None:
            continue

        proportion = required / total_required

        allocations[goal["name"]] = (
            money_for_goals * proportion
        )

    return {
        "allocations": allocations,
        "extra_money": 0,
        "shortfall": total_required - money_for_goals
    }


def use_extra_money_for_priority(
    goals,
    allocations,
    extra_money
):
    """
    Use extra saving money on higher-priority goals first.
    """

    priority_order = {
        "Must have": 1,
        "Important": 2,
        "Nice to have": 3
    }

    active_goals = [
        goal for goal in goals
        if goal.get("status") == "active"
    ]

    active_goals.sort(
        key=lambda goal: priority_order.get(
            goal.get("priority"),
            99
        )
    )

    updated_allocations = allocations.copy()

    for goal in active_goals:

        if extra_money <= 0:
            break

        already_allocated = updated_allocations.get(
            goal["name"],
            0
        )

        remaining_after_minimum = max(
            0,
            goal["remaining_amount"] - already_allocated
        )

        extra_for_goal = min(
            extra_money,
            remaining_after_minimum
        )

        updated_allocations[goal["name"]] = (
            already_allocated + extra_for_goal
        )

        extra_money -= extra_for_goal

    return updated_allocations, extra_money
````

````python
def apply_goal_allocations(goals, allocations):
    """
    Apply allocated savings to each active goal.
    Prevent a goal from being funded above its target.
    """

    messages = []
    unused_money = 0

    for goal in goals:

        allocation = allocations.get(
            goal["name"],
            0
        )

        if allocation < 0:
            raise ValueError(
                "Goal allocation cannot be negative."
            )

        if goal.get("status") != "active":
            unused_money += allocation
            continue

        remaining_amount = max(
            goal["target_amount"] - goal["current_saved"],
            0
        )

        amount_used = min(
            allocation,
            remaining_amount
        )

        goal["current_saved"] += amount_used

        unused_money += (
            allocation - amount_used
        )

        if goal["current_saved"] >= goal["target_amount"]:
            goal["current_saved"] = goal["target_amount"]
            goal["status"] = "completed"

            messages.append(
                f"Congratulations! You reached your "
                f"{goal['name']} goal."
            )

        else:
            remaining = (
                goal["target_amount"]
                - goal["current_saved"]
            )

            messages.append(
                f"{goal['name']}: "
                f"${remaining:.2f} remaining."
            )

    return goals, messages, unused_money
````

````python
def add_goal(
    goals,
    name,
    target_amount,
    deadline,
    priority,
    current_saved=0
):
    """
    Add a new saving goal.
    """

    if target_amount <= 0:
        raise ValueError("Target amount must be greater than 0.")

    if current_saved < 0:
        raise ValueError("Current savings cannot be negative.")

    new_goal = {
        "name": name,
        "target_amount": target_amount,
        "deadline": deadline,
        "priority": priority,
        "current_saved": current_saved,
        "status": "active"
    }

    goals.append(new_goal)

    return goals


def edit_goal(
    goals,
    goal_name,
    target_amount=None,
    deadline=None,
    priority=None
):
    """
    Edit an existing saving goal.
    """

    for goal in goals:

        if goal["name"].lower() == goal_name.lower():

            if target_amount is not None:
                if target_amount <= 0:
                    raise ValueError(
                        "Target amount must be greater than 0."
                    )

                goal["target_amount"] = target_amount

            if deadline is not None:
                goal["deadline"] = deadline

            if priority is not None:
                goal["priority"] = priority

            return goals

    raise ValueError("Goal not found.")


def change_goal_status(
    goals,
    goal_name,
    new_status
):
    """
    Pause or resume a saving goal.
    """

    if new_status not in ["active", "paused"]:
        raise ValueError(
            "Status must be active or paused."
        )

    for goal in goals:

        if goal["name"].lower() == goal_name.lower():
            goal["status"] = new_status
            return goals

    raise ValueError("Goal not found.")


def cancel_goal(
    goals,
    goal_name
):
    """
    Cancel a goal and release the money already saved for it.
    """

    for goal in goals:

        if goal["name"].lower() == goal_name.lower():

            released_money = goal.get(
                "current_saved",
                0
            )

            goal["current_saved"] = 0
            goal["status"] = "cancelled"

            return goals, released_money

    raise ValueError("Goal not found.")
````

## 5F – Planning and Replanning

If the current saving plan is not achievable, the application can look for flexible expenses that may be reduced and suggest changes before asking the user to adjust a saving goal.

````python
def find_expense_reduction_options(
    expenses,
    shortfall,
    budget_period="weekly"
):
    """
    Find optional flexible expenses that could
    help cover a saving-plan shortfall.
    """

    if shortfall < 0:
        raise ValueError(
            "Shortfall cannot be negative."
        )

    options = []

    for expense in expenses:

        if expense.get("status") != "confirmed":
            continue

        if expense.get("classification") != "optional flexible":
            continue

        user_amount = calculate_user_share(
            expense["amount"],
            expense.get("share_percent", 100)
        )

        current_amount = normalise_frequency(
            user_amount,
            expense["frequency"],
            budget_period
        )

        possible_reduction = min(
            current_amount,
            shortfall
        )

        options.append({
            "name": expense["name"],
            "current_amount": current_amount,
            "possible_reduction": possible_reduction,
            "suggested_amount":
                current_amount - possible_reduction
        })

    options.sort(
        key=lambda item: item["current_amount"],
        reverse=True
    )

    return options


def apply_expense_reduction(
    expenses,
    expense_name,
    reduction_amount,
    budget_period="weekly"
):
    """
    Apply an accepted expense reduction while
    keeping the expense's original frequency.
    """

    if reduction_amount < 0:
        raise ValueError(
            "Reduction amount cannot be negative."
        )

    for expense in expenses:

        if expense["name"].lower() == expense_name.lower():

            user_amount = calculate_user_share(
                expense["amount"],
                expense.get("share_percent", 100)
            )

            current_amount = normalise_frequency(
                user_amount,
                expense["frequency"],
                budget_period
            )

            if reduction_amount > current_amount:
                raise ValueError(
                    "Reduction cannot exceed the expense."
                )

            if current_amount == 0:
                expense["amount"] = 0

            else:
                remaining_ratio = (
                    current_amount - reduction_amount
                ) / current_amount

                expense["amount"] *= remaining_ratio

            return expenses

    raise ValueError("Expense not found.")
````

````python
def find_adjustable_goal(goals):
    """
    Find the lowest-priority active goal that can be adjusted.
    Must-have goals are protected.
    """

    priority_order = {
        "Nice to have": 1,
        "Important": 2,
        "Must have": 3
    }

    adjustable_goals = [
        goal for goal in goals
        if (
            goal.get("status") == "active"
            and goal.get("priority") != "Must have"
        )
    ]

    if not adjustable_goals:
        return None

    adjustable_goals.sort(
        key=lambda goal: priority_order.get(
            goal.get("priority"),
            99
        )
    )

    return adjustable_goals[0]


def suggest_deadline_extension(goal, shortfall):
    """
    Suggest a longer saving period for a goal
    to reduce the weekly shortfall.
    """

    if shortfall <= 0:
        return None

    required_now = goal.get("required_per_period")
    periods_left = goal.get("periods_left")
    remaining_amount = goal.get("remaining_amount")

    if (
        required_now is None
        or periods_left is None
        or remaining_amount is None
        or required_now <= 0
    ):
        return None

    new_required = max(
        required_now - shortfall,
        0
    )

    if new_required <= 0:
        return None

    new_periods = math.ceil(
        remaining_amount / new_required
    )

    if new_periods <= periods_left:
        return None

    return {
        "goal": goal["name"],
        "current_periods": periods_left,
        "suggested_periods": new_periods,
        "new_required_per_period":
            remaining_amount / new_periods
    }
````

````python
def replan_from_state(state, reference_date=None):
    """
    Recalculate the complete financial plan
    using the latest saved user data.
    """

    state_copy = copy.deepcopy(state)

    # Income
    general_income, restricted_income = calculate_income_totals(
        state_copy.get("income", []),
        budget_period="weekly"
    )

    # Expenses
    expenses = clean_expense_data(
        state_copy.get("expenses", [])
    )

    total_expenses = calculate_total_expenses(
        expenses,
        budget_period="weekly"
    )

    money_remaining = calculate_money_remaining(
        general_income,
        total_expenses
    )

    # Buffer
    buffer_amount = state_copy.get("buffer", 0)

    if buffer_amount < 0:
        raise ValueError(
            "Buffer amount cannot be negative."
        )

    money_for_goals = calculate_money_for_goals(
        money_remaining,
        buffer_amount
    )

    # Goals
    goals = clean_goal_data(
        state_copy.get("goals", [])
    )

    for goal in goals:

        if (
            goal.get("status") == "active"
            and goal.get("deadline")
        ):
            goal["periods_left"] = calculate_weeks_left(
                goal["deadline"],
                reference_date
            )

    goals = update_goal_requirements(goals)

    # Feasibility
    plan_check = check_updated_plan(
        goals,
        money_for_goals
    )

    # Minimum allocations
    allocation_result = allocate_goal_savings(
        goals,
        money_for_goals
    )

    # Extra money goes to higher-priority goals
    final_allocations, unallocated_money = (
        use_extra_money_for_priority(
            goals,
            allocation_result["allocations"],
            allocation_result["extra_money"]
        )
    )

    return {
        "general_income": general_income,
        "restricted_income": restricted_income,
        "total_expenses": total_expenses,
        "money_remaining": money_remaining,
        "buffer": buffer_amount,
        "money_for_goals": money_for_goals,
        "goals": goals,
        "status": plan_check["status"],
        "total_required": plan_check["total_required"],
        "surplus": plan_check["surplus"],
        "shortfall": plan_check["shortfall"],
        "overdue_goals": plan_check["overdue_goals"],
        "allocations": final_allocations,
        "unallocated_money": unallocated_money
    }
````

## 5G – Gemini AI Functions

Gemini is used for natural-language understanding and conversational explanations. Python remains responsible for the actual financial calculations so that the results are consistent and testable.

````python
FINANCE_PERSONA = """
You are a friendly personal finance assistant.

Your job is to help users understand their own income, expenses,
saving goals and spending patterns.

Use the user's actual financial data when it is provided.

Do not invent financial numbers or change results calculated by Python.

Never assume a missing time period or frequency.
If the user does not say weekly, monthly or yearly, do not guess it.

Keep explanations simple and practical.

If a question is unrelated to personal finance, politely explain
that your role is focused on personal finance.
"""


def ask_gemini(prompt, system_instruction=FINANCE_PERSONA):
    """
    Send a prompt to Gemini using the REST API.
    Retry temporary API errors and return a friendly
    message if Gemini is unavailable.
    """

    payload = {
        "system_instruction": {
            "parts": [
                {
                    "text": system_instruction
                }
            ]
        },
        "contents": [
            {
                "parts": [
                    {
                        "text": prompt
                    }
                ]
            }
        ]
    }

    temporary_errors = [
        408, 429, 500, 502, 503, 504
    ]

    for attempt in range(3):

        try:
            response = requests.post(
                GEMINI_URL,
                params={
                    "key": GEMINI_API_KEY
                },
                json=payload,
                timeout=30
            )

            if response.status_code == 200:

                result = response.json()

                return (
                    result["candidates"][0]
                    ["content"]["parts"][0]["text"]
                )

            if response.status_code in temporary_errors:
                time.sleep(2)
                continue

            return (
                "Gemini request failed. "
                "Please try again later."
            )

        except requests.exceptions.RequestException:

            if attempt < 2:
                time.sleep(2)

    return (
        "Gemini is temporarily unavailable. "
        "The finance calculations can still be used."
    )
````

## 5H – Spending Tracking and CSV

The application can record actual transactions and compare weekly spending with the user's planned budget. Transactions can be entered manually or loaded from a CSV file.

````python
def add_transaction(
    transactions,
    date_value,
    category,
    amount,
    description=""
):
    """
    Add a spending transaction.
    """

    if amount < 0:
        raise ValueError(
            "Transaction amount cannot be negative."
        )

    # Check date format
    datetime.strptime(
        date_value,
        "%Y-%m-%d"
    )

    transactions.append({
        "date": date_value,
        "category": category,
        "amount": amount,
        "description": description
    })

    return transactions


def filter_transactions_by_week(
    transactions,
    reference_date=None
):
    """
    Return transactions from the Monday-Sunday week
    containing the reference date.
    """

    if reference_date is None:
        reference_date = date.today()

    if isinstance(reference_date, str):
        reference_date = datetime.strptime(
            reference_date,
            "%Y-%m-%d"
        ).date()

    week_start = (
        reference_date
        - timedelta(days=reference_date.weekday())
    )

    week_end = week_start + timedelta(days=6)

    weekly_transactions = []

    for transaction in transactions:

        transaction_date = datetime.strptime(
            transaction["date"],
            "%Y-%m-%d"
        ).date()

        if week_start <= transaction_date <= week_end:
            weekly_transactions.append(
                transaction
            )

    return weekly_transactions
````

````python
def normalise_expense_name(name):
    """
    Standardise similar expense names for comparison.
    """

    name = name.strip().lower()

    aliases = {
        "food": "groceries",
        "grocery": "groceries",
        "mobile": "phone",
        "mobile phone": "phone",
        "phone bill": "phone",
        "public transport": "transport",
        "bus": "transport",
        "train": "transport",
        "transportation": "transport",
        "eating out": "going out",
        "entertainment": "going out"
    }

    return aliases.get(name, name)


def compare_actual_to_plan(
    transactions,
    planned_expenses,
    budget_period="weekly",
    period_complete=False
):
    """
    Compare actual spending with planned spending.
    """

    actual_totals = {}

    for transaction in transactions:

        category = normalise_expense_name(
            transaction["category"]
        )

        actual_totals[category] = (
            actual_totals.get(category, 0)
            + transaction["amount"]
        )

    comparison = []

    for expense in planned_expenses:

        if expense.get("status", "confirmed") != "confirmed":
            continue

        category = normalise_expense_name(
            expense["name"]
        )

        user_share = calculate_user_share(
            expense["amount"],
            expense.get("share_percent", 100)
        )

        planned_amount = normalise_frequency(
            user_share,
            expense["frequency"],
            budget_period
        )

        actual_amount = actual_totals.get(
            category,
            0
        )

        difference = actual_amount - planned_amount

        if difference > 0:
            status = "over budget"

        elif difference == 0:
            status = "on budget"

        elif period_complete:
            status = "under budget"

        else:
            status = "within budget so far"

        comparison.append({
            "category": expense["name"],
            "planned": planned_amount,
            "actual": actual_amount,
            "difference": difference,
            "status": status
        })

    return comparison
````

````python
def calculate_overspending_impact(
    spending_comparison,
    normal_money_for_goals,
    goals
):
    """
    Calculate how overspending reduces the money
    available for saving goals.
    """

    total_overspending = 0

    for item in spending_comparison:

        if item["difference"] > 0:
            total_overspending += item["difference"]

    adjusted_money_for_goals = max(
        0,
        normal_money_for_goals - total_overspending
    )

    updated_plan = check_updated_plan(
        goals,
        adjusted_money_for_goals
    )

    return {
        "total_overspending": total_overspending,
        "money_for_goals_after_overspending":
            adjusted_money_for_goals,
        "status": updated_plan["status"],
        "surplus": updated_plan["surplus"],
        "shortfall": updated_plan["shortfall"]
    }


def create_weekly_feedback(
    spending_comparison,
    impact
):
    """
    Create verified weekly feedback using
    Python-calculated results.
    """

    messages = []

    for item in spending_comparison:

        if item["difference"] > 0:

            messages.append(
                f"{item['category']} was over budget by "
                f"${item['difference']:.2f}."
            )

    if impact["total_overspending"] > 0:

        messages.append(
            f"Total overspending this week was "
            f"${impact['total_overspending']:.2f}."
        )

    else:
        messages.append(
            "There was no overspending this week."
        )

    if impact["status"] == "achievable":

        messages.append(
            "Your current saving goals are still achievable."
        )

    else:

        messages.append(
            f"Your saving plan now has a shortfall of "
            f"${impact['shortfall']:.2f}."
        )

    return messages
````

````python
def load_transactions_csv(file_path):
    """
    Load and clean transaction data from a CSV file.
    """

    df = pd.read_csv(file_path)

    required_columns = [
        "Date",
        "Category",
        "Amount"
    ]

    for column in required_columns:
        if column not in df.columns:
            raise ValueError(
                f"Missing required column: {column}"
            )

    # Description is optional
    if "Description" not in df.columns:
        df["Description"] = ""

    # Clean amount and date
    df["Amount"] = pd.to_numeric(
        df["Amount"],
        errors="coerce"
    )

    df["Date"] = pd.to_datetime(
        df["Date"],
        errors="coerce"
    )

    # Remove rows that cannot be used
    df = df.dropna(
        subset=["Date", "Category", "Amount"]
    )

    transactions = []

    for _, row in df.iterrows():

        category = str(row["Category"]).strip()

        if category == "":
            continue

        transactions.append({
            "date": row["Date"].strftime("%Y-%m-%d"),
            "category": category,
            "amount": float(row["Amount"]),
            "description": str(row["Description"])
        })

    return transactions
````

````python
def generate_weekly_advice(feedback_messages):
    """
    Ask Gemini to turn verified Python results
    into a short weekly finance summary.
    """

    verified_facts = "\n".join(
        feedback_messages
    )

    prompt = f"""
Turn the verified financial facts below into a short,
friendly weekly summary.

Rules:
- Use only the facts provided.
- Do not invent numbers.
- Do not add new financial conclusions.
- Do not strengthen or weaken the verified conclusions.
- If the facts say "achievable", say "achievable",
  not "completely", "definitely", or "guaranteed".
- Keep the response short and practical.

Verified facts:
{verified_facts}
"""

    reply = ask_gemini(prompt)

    if reply.startswith("Gemini"):
        return "\n".join(feedback_messages)

    return reply
````

## 6D – Bug Found and Fixed

During testing, I found that the spending comparison originally labelled a category as "under budget" even when the week was not finished.

This could be misleading because the user may still spend more later in the week.

I changed the logic so that:
- during an unfinished week, spending below the plan is shown as **"within budget so far"**
- only when the period is complete is it shown as **"under budget"**

This was a useful example of checking the business meaning of the output rather than only checking whether the code ran.

````python
# =========================================================
# STEP 7C - UI FUNCTIONS
# =========================================================


# ---------------------------------------------------------
# GENERAL HELPERS
# ---------------------------------------------------------

def is_missing(value):
    if value is None:
        return True

    try:
        if pd.isna(value):
            return True
    except:
        pass

    return str(value).strip() == ""


def table_to_records(data):
    if data is None:
        return []

    df = pd.DataFrame(data)
    df = df.where(pd.notna(df), None)

    return df.to_dict("records")


def get_names(data):
    names = []

    for row in table_to_records(data):
        name = row.get("name")

        if not is_missing(name):
            names.append(str(name))

    return names


# ---------------------------------------------------------
# PROFILE
# ---------------------------------------------------------

def build_profile_expense_checklist(profile):

    suggestions = [
        "Groceries",
        "Phone",
        "Transport"
    ]

    if profile.get("housing_cost") == "Yes":

        suggestions.append(
            "Rent or mortgage"
        )

        if profile.get("housing_bills") in [
            "Partly included",
            "Not included",
            "Not sure"
        ]:
            suggestions.extend([
                "Electricity",
                "Water",
                "Gas",
                "Internet"
            ])

    if profile.get("role") == "Student":
        suggestions.extend([
            "Study materials",
            "Course costs",
            "Software subscriptions"
        ])

    if profile.get("car") == "Yes":
        suggestions.extend([
            "Fuel",
            "Car insurance",
            "Registration",
            "Servicing"
        ])

    if profile.get("children") == "Yes":
        suggestions.extend([
            "Childcare or school costs",
            "Children's clothing",
            "Children's activities"
        ])

    if profile.get("pets") == "Yes":
        suggestions.extend([
            "Pet food",
            "Vet costs",
            "Pet insurance"
        ])

    if profile.get("dependants") == "Yes":
        suggestions.extend([
            "Family support",
            "Care costs"
        ])

    if profile.get("hobbies"):
        suggestions.append(
            "Hobbies and activities"
        )

    return list(
        dict.fromkeys(suggestions)
    )


def ui_save_profile(
    name,
    role,
    household_size,
    living_situation,
    housing_cost,
    housing_bills,
    car,
    pets,
    children,
    children_count,
    child_age_group,
    dependants,
    hobbies
):

    if is_missing(name):
        raise gr.Error(
            "Please enter your name or nickname."
        )

    if is_missing(role):
        raise gr.Error(
            "Please choose what best describes you."
        )

    if (
        household_size is None
        or household_size < 1
    ):
        raise gr.Error(
            "Please enter your household size."
        )

    if is_missing(living_situation):
        raise gr.Error(
            "Please choose your living situation."
        )

    if is_missing(housing_cost):
        raise gr.Error(
            "Please answer the housing question."
        )

    if (
        housing_cost == "Yes"
        and is_missing(housing_bills)
    ):
        raise gr.Error(
            "Please tell us whether household bills are included."
        )

    if is_missing(car):
        raise gr.Error(
            "Please answer the car question."
        )

    if is_missing(pets):
        raise gr.Error(
            "Please answer the pets question."
        )

    if is_missing(children):
        raise gr.Error(
            "Please answer the children question."
        )

    if children == "Yes":

        if (
            children_count is None
            or children_count < 1
        ):
            raise gr.Error(
                "Please enter the number of children."
            )

        if is_missing(child_age_group):
            raise gr.Error(
                "Please select the child age group."
            )

    if is_missing(dependants):
        raise gr.Error(
            "Please answer the dependants question."
        )

    profile = {
        "name": name.strip(),
        "role": role,
        "household_size": int(
            household_size
        ),
        "living_situation": living_situation,
        "housing_cost": housing_cost,
        "housing_bills": (
            housing_bills
            if housing_cost == "Yes"
            else "Not applicable"
        ),
        "car": car,
        "pets": pets,
        "children": children,
        "children_count": (
            int(children_count)
            if children == "Yes"
            else 0
        ),
        "child_age_group": (
            child_age_group
            if children == "Yes"
            else None
        ),
        "dependants": dependants,
        "hobbies": (
            hobbies.strip()
            if hobbies else ""
        )
    }

    suggestions = (
        build_profile_expense_checklist(
            profile
        )
    )

    message = (
        f"### ✅ Profile saved\n\n"
        f"Hi **{profile['name']}** 🌸\n\n"
        "Based on your situation, you may also want "
        "to check these expenses:\n\n"
    )

    for item in suggestions:
        message += f"- {item}\n"

    message += (
        "\nThese are only reminders. "
        "**Nothing has been added automatically.**"
    )

    gr.Info(
        f"✅ Profile saved. Hi {profile['name']}!"
    )

    return profile, message


# ---------------------------------------------------------
# AI EXTRACTION
# ---------------------------------------------------------

def ui_extract_income(
    text,
    profile
):

    if not text or not text.strip():
        raise gr.Error(
            "Please describe your income first."
        )

    data = extract_income_with_gemini(
        text
    )

    if not data:
        raise gr.Error(
            "I could not understand the income information."
        )

    # If the user says "I / me / my",
    # use the name saved in their profile.
    user_name = ""

    if profile:
        user_name = str(
            profile.get("name", "")
        ).strip()

    generic_self_names = {
        "user",
        "user income",
        "my income",
        "me",
        "i",
        "self",
        "income"
    }

    if user_name:

        for row in data:

            source_name = str(
                row.get("name", "")
            ).strip().lower()

            if source_name in generic_self_names:
                row["name"] = user_name

    df = pd.DataFrame(data)
    names = get_names(df)

    gr.Info(
        "✅ Income extracted successfully."
    )

    status = """
### ✅ Income extracted

Review the table below ↓

If something is missing or incorrect,
open **Something missing or wrong? Fix it here**.
"""

    return (
        df,
        status,
        gr.update(
            choices=names,
            value=(
                names[0]
                if names
                else None
            )
        )
    )


def ui_extract_expenses(text):

    if not text or not text.strip():
        raise gr.Error(
            "Please describe your expenses first."
        )

    data = extract_expenses_with_gemini(
        text
    )

    if not data:
        raise gr.Error(
            "I could not understand the expense information."
        )

    data = clean_expense_data(
        data
    )

    df = pd.DataFrame(data)
    names = get_names(df)

    gr.Info(
        "✅ Expenses extracted successfully."
    )

    status = """
### ✅ Expenses extracted

Review the table below ↓

If something is missing or incorrect,
open **Something missing or wrong? Fix it here**.
"""

    return (
        df,
        status,
        gr.update(
            choices=names,
            value=(
                names[0]
                if names
                else None
            )
        )
    )


def ui_extract_goals(text):

    if not text or not text.strip():
        raise gr.Error(
            "Please describe your saving goals first."
        )

    data = extract_goals_with_gemini(
        text
    )

    if not data:
        raise gr.Error(
            "I could not understand the saving goals."
        )

    data = clean_goal_data(
        data
    )

    # AUD-only app
    for goal in data:
        goal.pop(
            "currency",
            None
        )

    df = pd.DataFrame(data)
    names = get_names(df)

    gr.Info(
        "✅ Saving goals extracted successfully."
    )

    status = """
### ✅ Saving goals extracted

Review the table below ↓

A blank **deadline** or **priority** is intentional if
you did not provide that information. The AI will not guess it.
"""

    return (
        df,
        status,
        gr.update(
            choices=names,
            value=(
                names[0]
                if names
                else None
            )
        )
    )


# ---------------------------------------------------------
# LOAD CURRENT VALUES INTO CORRECTION FORM
# ---------------------------------------------------------

def ui_load_income_for_edit(
    data,
    item_name
):

    for row in table_to_records(data):

        if str(row.get("name")) == str(
            item_name
        ):

            return (
                row.get("amount"),
                row.get("frequency"),
                row.get(
                    "availability_type"
                ),
                row.get(
                    "availability_value"
                )
            )

    return None, None, None, None


def ui_load_expense_for_edit(
    data,
    item_name
):

    for row in table_to_records(data):

        if str(row.get("name")) == str(
            item_name
        ):

            return (
                row.get("amount"),
                row.get("frequency"),
                row.get(
                    "classification"
                )
            )

    return None, None, None


def ui_load_goal_for_edit(
    data,
    item_name
):

    for row in table_to_records(data):

        if str(row.get("name")) == str(
            item_name
        ):

            return (
                row.get(
                    "target_amount"
                ),
                row.get("deadline"),
                row.get("priority"),
                row.get(
                    "current_saved",
                    0
                )
            )

    return None, None, None, None


# ---------------------------------------------------------
# UPDATE EXTRACTED INFORMATION
# ---------------------------------------------------------

def ui_update_income(
    data,
    item_name,
    amount,
    frequency,
    availability,
    availability_value
):

    rows = table_to_records(data)

    if not rows:
        raise gr.Error(
            "Extract income first."
        )

    if is_missing(item_name):
        raise gr.Error(
            "Choose an income source."
        )

    found = False

    for row in rows:

        if str(row.get("name")) == str(
            item_name
        ):

            found = True

            if amount is not None:

                if amount < 0:
                    raise gr.Error(
                        "Income cannot be negative."
                    )

                row["amount"] = float(
                    amount
                )

            if frequency:
                row["frequency"] = frequency

            if availability:

                row[
                    "availability_type"
                ] = availability

                if availability in [
                    "all",
                    "none"
                ]:
                    row[
                        "availability_value"
                    ] = None

            current_type = row.get(
                "availability_type"
            )

            if (
                availability_value
                is not None
                and current_type in [
                    "fixed",
                    "percent"
                ]
            ):
                row[
                    "availability_value"
                ] = float(
                    availability_value
                )

    if not found:
        raise gr.Error(
            "Income source not found."
        )

    gr.Info(
        "✅ Income updated."
    )

    return pd.DataFrame(rows)


def ui_update_expense(
    data,
    item_name,
    amount,
    frequency,
    classification
):

    rows = table_to_records(data)

    if not rows:
        raise gr.Error(
            "Extract expenses first."
        )

    if is_missing(item_name):
        raise gr.Error(
            "Choose an expense."
        )

    found = False

    for row in rows:

        if str(row.get("name")) == str(
            item_name
        ):

            found = True

            if amount is not None:

                if amount < 0:
                    raise gr.Error(
                        "Expense cannot be negative."
                    )

                row["amount"] = float(
                    amount
                )

            if frequency:
                row["frequency"] = frequency

            if classification:
                row[
                    "classification"
                ] = classification

    if not found:
        raise gr.Error(
            "Expense not found."
        )

    gr.Info(
        "✅ Expense updated."
    )

    return pd.DataFrame(rows)


def ui_update_goal(
    data,
    item_name,
    target,
    deadline,
    priority,
    current_saved
):

    rows = table_to_records(data)

    if not rows:
        raise gr.Error(
            "Extract saving goals first."
        )

    if is_missing(item_name):
        raise gr.Error(
            "Choose a saving goal."
        )

    found = False

    for row in rows:

        if str(row.get("name")) == str(
            item_name
        ):

            found = True

            if target is not None:

                if target <= 0:
                    raise gr.Error(
                        "Target must be greater than 0."
                    )

                row[
                    "target_amount"
                ] = float(target)

            if (
                deadline
                and deadline.strip()
            ):

                try:
                    datetime.strptime(
                        deadline.strip(),
                        "%Y-%m-%d"
                    )

                except ValueError:
                    raise gr.Error(
                        "Deadline must use YYYY-MM-DD."
                    )

                row[
                    "deadline"
                ] = deadline.strip()

            if priority:
                row[
                    "priority"
                ] = priority

            if current_saved is not None:

                if current_saved < 0:
                    raise gr.Error(
                        "Current savings cannot be negative."
                    )

                row[
                    "current_saved"
                ] = float(
                    current_saved
                )

            row.pop(
                "currency",
                None
            )

    if not found:
        raise gr.Error(
            "Saving goal not found."
        )

    gr.Info(
        "✅ Saving goal updated."
    )

    return pd.DataFrame(rows)


# ---------------------------------------------------------
# FINAL CHECK
# ---------------------------------------------------------

def collect_required_issues(
    profile,
    income_data,
    expense_data,
    goal_data
):

    problems = []

    incomes = table_to_records(
        income_data
    )

    expenses = table_to_records(
        expense_data
    )

    goals = table_to_records(
        goal_data
    )

    if not profile:
        problems.append(
            "About You: save your profile first."
        )

    if not incomes:
        problems.append(
            "Income: add at least one income source."
        )

    if not expenses:
        problems.append(
            "Expenses: add your expense information."
        )

    if not goals:
        problems.append(
            "Goals: add at least one saving goal."
        )

    allowed_frequencies = {
        "weekly",
        "fortnightly",
        "monthly",
        "yearly"
    }

    # Income validation
    for row in incomes:

        name = row.get(
            "name",
            "Income"
        )

        if is_missing(row.get("amount")):
            problems.append(
                f"{name}: amount is missing."
            )

        frequency = str(
            row.get(
                "frequency",
                ""
            )
        ).lower()

        if (
            frequency
            not in allowed_frequencies
        ):
            problems.append(
                f"{name}: choose a valid frequency."
            )

        availability = str(
            row.get(
                "availability_type",
                ""
            )
        ).lower()

        if availability not in {
            "all",
            "fixed",
            "percent",
            "none"
        }:
            problems.append(
                f"{name}: availability is missing."
            )

        if (
            availability
            in {"fixed", "percent"}
            and is_missing(
                row.get(
                    "availability_value"
                )
            )
        ):
            problems.append(
                f"{name}: available amount or percentage is missing."
            )

    # Expense validation
    for row in expenses:

        name = row.get(
            "name",
            "Expense"
        )

        if is_missing(
            row.get("amount")
        ):
            problems.append(
                f"{name}: amount is missing."
            )

        frequency = str(
            row.get(
                "frequency",
                ""
            )
        ).lower()

        if (
            frequency
            not in allowed_frequencies
        ):
            problems.append(
                f"{name}: choose a valid frequency."
            )

        classification = str(
            row.get(
                "classification",
                ""
            )
        ).lower()

        if classification not in {
            "fixed",
            "essential flexible",
            "optional flexible"
        }:
            problems.append(
                f"{name}: choose a valid expense type."
            )

    # Goal validation
    for row in goals:

        name = row.get(
            "name",
            "Goal"
        )

        if is_missing(
            row.get(
                "target_amount"
            )
        ):
            problems.append(
                f"{name}: target amount is missing."
            )

        if is_missing(
            row.get("deadline")
        ):
            problems.append(
                f"{name}: deadline is missing."
            )

        else:

            try:
                datetime.strptime(
                    str(
                        row["deadline"]
                    ),
                    "%Y-%m-%d"
                )

            except ValueError:
                problems.append(
                    f"{name}: deadline must be YYYY-MM-DD."
                )

        priority = str(
            row.get(
                "priority",
                ""
            )
        ).lower()

        if priority not in {
            "must have",
            "important",
            "nice to have"
        }:
            problems.append(
                f"{name}: choose a priority."
            )

    return problems


def ui_check_required_information(
    profile,
    income_data,
    expense_data,
    goal_data
):

    problems = collect_required_issues(
        profile,
        income_data,
        expense_data,
        goal_data
    )

    if problems:

        message = (
            "### ⚠️ Still missing information\n\n"
        )

        for problem in problems:
            message += (
                f"- {problem}\n"
            )

        message += (
            "\nOpen **Something missing or wrong? "
            "Fix it here** above to complete these fields."
        )

        return message

    return """
### ✅ Ready to Create Your Plan

Everything required has been completed. 🌸
"""


# ---------------------------------------------------------
# CREATE PLAN + PLAN TABLES
# ---------------------------------------------------------

def build_plan_suggestions(plan, expenses):
    """
    Create simple practical suggestions from the calculated plan.
    """

    rows = []

    status = plan.get("status")
    surplus = float(plan.get("surplus", 0) or 0)
    shortfall = float(plan.get("shortfall", 0) or 0)

    # -----------------------------------------------------
    # PLAN IS ACHIEVABLE
    # -----------------------------------------------------

    if status == "achievable":

        if surplus > 0:
            rows.append([
                "Extra savings",
                "Priority goals",
                f"${surplus:.2f} available",
                "Put extra money toward higher-priority goals",
                "All minimum goal requirements are already covered."
            ])

        else:
            rows.append([
                "No adjustment needed",
                "-",
                "-",
                "-",
                "Your current income and expenses can support your goals."
            ])

        return pd.DataFrame(
            rows,
            columns=[
                "Suggestion",
                "Item",
                "Current",
                "Suggested",
                "Reason"
            ]
        )

    # -----------------------------------------------------
    # OVERDUE GOALS
    # -----------------------------------------------------

    for goal_name in plan.get(
        "overdue_goals",
        []
    ):
        rows.append([
            "Review deadline",
            goal_name,
            "Deadline has passed",
            "Choose a new realistic deadline",
            "The current deadline cannot be used for a future saving plan."
        ])

    # -----------------------------------------------------
    # OPTIONAL EXPENSE REDUCTIONS
    # -----------------------------------------------------

    remaining_shortfall = shortfall

    for expense in expenses:

        if remaining_shortfall <= 0:
            break

        if (
            expense.get("status", "confirmed")
            != "confirmed"
        ):
            continue

        if (
            str(
                expense.get(
                    "classification",
                    ""
                )
            ).lower()
            != "optional flexible"
        ):
            continue

        try:
            user_amount = calculate_user_share(
                float(expense["amount"]),
                float(
                    expense.get(
                        "share_percent",
                        100
                    )
                )
            )

            weekly_amount = normalise_frequency(
                user_amount,
                expense["frequency"],
                "weekly"
            )

        except (ValueError, TypeError, KeyError):
            continue

        reduction = min(
            weekly_amount,
            remaining_shortfall
        )

        suggested_amount = max(
            0,
            weekly_amount - reduction
        )

        if reduction > 0:

            rows.append([
                "Reduce optional spending",
                expense.get(
                    "name",
                    "Expense"
                ),
                f"${weekly_amount:.2f}/week",
                f"${suggested_amount:.2f}/week",
                (
                    f"Reducing this by ${reduction:.2f}/week "
                    "would help close the goal shortfall."
                )
            ])

            remaining_shortfall -= reduction

    # -----------------------------------------------------
    # LOWER-PRIORITY DEADLINE
    # -----------------------------------------------------

    try:
        adjustable_goal = find_adjustable_goal(
            plan["goals"]
        )

        if adjustable_goal is not None:

            deadline_option = suggest_deadline_extension(
                adjustable_goal,
                shortfall
            )

            if deadline_option is not None:

                rows.append([
                    "Extend goal deadline",
                    deadline_option["goal"],
                    (
                        f"{deadline_option['current_periods']} "
                        "weeks remaining"
                    ),
                    (
                        f"About {deadline_option['suggested_periods']} "
                        "weeks"
                    ),
                    (
                        "This could reduce the required saving to "
                        f"${deadline_option['new_required_per_period']:.2f}"
                        "/week."
                    )
                ])

    except (ValueError, KeyError, TypeError):
        pass

    # -----------------------------------------------------
    # FALLBACK
    # -----------------------------------------------------

    if not rows:

        rows.append([
            "Review your plan",
            "-",
            f"${shortfall:.2f} shortfall/week",
            "Adjust spending or a lower-priority goal",
            "Your current plan does not have enough weekly money."
        ])

    return pd.DataFrame(
        rows,
        columns=[
            "Suggestion",
            "Item",
            "Current",
            "Suggested",
            "Reason"
        ]
    )


def ui_create_plan(
    profile,
    income_data,
    expense_data,
    goal_data,
    buffer_amount
):

    problems = collect_required_issues(
        profile,
        income_data,
        expense_data,
        goal_data
    )

    if problems:
        raise gr.Error(
            "Please complete the required information "
            "before creating your plan."
        )

    incomes = table_to_records(
        income_data
    )

    expenses = table_to_records(
        expense_data
    )

    goals = table_to_records(
        goal_data
    )

    # -----------------------------------------------------
    # CLEAN INCOME
    # -----------------------------------------------------

    for row in incomes:

        row["amount"] = float(
            row["amount"]
        )

        row["frequency"] = str(
            row["frequency"]
        ).lower()

        row["availability_type"] = str(
            row["availability_type"]
        ).lower()

        if row["availability_type"] in [
            "fixed",
            "percent"
        ]:
            row["availability_value"] = float(
                row["availability_value"]
            )

        else:
            row["availability_value"] = None

    # -----------------------------------------------------
    # CLEAN EXPENSES
    # -----------------------------------------------------

    expenses = clean_expense_data(
        expenses
    )

    for row in expenses:

        row["amount"] = float(
            row["amount"]
        )

        row["frequency"] = str(
            row["frequency"]
        ).lower()

        row["classification"] = str(
            row["classification"]
        ).lower()

    # -----------------------------------------------------
    # CLEAN GOALS
    # -----------------------------------------------------

    goals = clean_goal_data(
        goals
    )

    priority_names = {
        "must have": "Must have",
        "important": "Important",
        "nice to have": "Nice to have"
    }

    for row in goals:

        row["target_amount"] = float(
            row["target_amount"]
        )

        row["priority"] = priority_names[
            str(
                row["priority"]
            ).lower()
        ]

        if is_missing(
            row.get(
                "current_saved"
            )
        ):
            row["current_saved"] = 0

        row["current_saved"] = float(
            row["current_saved"]
        )

        row.pop(
            "currency",
            None
        )

    # -----------------------------------------------------
    # BUFFER
    # -----------------------------------------------------

    if buffer_amount is None:
        buffer_amount = 0

    if buffer_amount < 0:
        raise gr.Error(
            "Safety buffer cannot be negative."
        )

    # -----------------------------------------------------
    # STATE
    # -----------------------------------------------------

    state = {
        "profile": profile,
        "income": incomes,
        "expenses": expenses,
        "goals": goals,
        "buffer": float(
            buffer_amount
        ),
        "transactions": []
    }

    plan = replan_from_state(
        state
    )

    # -----------------------------------------------------
    # STATUS
    # -----------------------------------------------------

    if plan["status"] == "achievable":
        status_text = "✅ Achievable"

    elif plan["status"] == "needs attention":
        status_text = "⚠️ Needs attention"

    else:
        status_text = "⚠️ Not achievable"

    # -----------------------------------------------------
    # TOP SUMMARY CARDS
    # -----------------------------------------------------

    income_card = f"""
### 💰 Weekly Income
**${plan['general_income']:.2f} AUD**
"""

    expense_card = f"""
### 🛍️ Weekly Expenses
**${plan['total_expenses']:.2f} AUD**
"""

    goal_card = f"""
### 🎯 Available for Goals
**${plan['money_for_goals']:.2f} AUD**
"""

    status_card = f"""
### 💗 Plan Status
**{status_text}**
"""

    # -----------------------------------------------------
    # STATUS MESSAGE
    # -----------------------------------------------------

    user_name = profile.get(
        "name",
        ""
    )

    if plan["status"] == "achievable":

        status_message = f"""
### ✅ {user_name}, your plan is achievable

Your minimum weekly goal requirements can be covered
with your current income, expenses and safety buffer.

See the tables below for the full breakdown.
"""

    else:

        status_message = f"""
### ⚠️ {user_name}, your plan needs adjustment

There is currently a weekly shortfall of
**${plan['shortfall']:.2f} AUD**.

Check the **Suggestions** table below for possible changes.
"""

    # -----------------------------------------------------
    # PLAN SUMMARY TABLE
    # -----------------------------------------------------

    summary_rows = [
        [
            "Weekly Income",
            f"${plan['general_income']:.2f} AUD"
        ],
        [
            "Weekly Expenses",
            f"${plan['total_expenses']:.2f} AUD"
        ],
        [
            "Safety Buffer",
            f"${plan['buffer']:.2f} AUD"
        ],
        [
            "Available for Goals",
            f"${plan['money_for_goals']:.2f} AUD"
        ],
        [
            "Minimum Needed for Goals",
            f"${plan['total_required']:.2f} AUD"
        ],
        [
            "Plan Status",
            status_text
        ]
    ]

    summary_df = pd.DataFrame(
        summary_rows,
        columns=[
            "Item",
            "Weekly Amount / Result"
        ]
    )

    # -----------------------------------------------------
    # GOAL ALLOCATION TABLE
    # -----------------------------------------------------

    goal_rows = []

    for goal in plan["goals"]:

        goal_name = goal.get(
            "name",
            ""
        )

        required = goal.get(
            "required_per_period"
        )

        if required is None:
            required_display = None
        else:
            required_display = round(
                float(required),
                2
            )

        allocation = plan[
            "allocations"
        ].get(
            goal_name,
            0
        )

        goal_rows.append([
            goal_name,
            round(
                float(
                    goal.get(
                        "target_amount",
                        0
                    )
                ),
                2
            ),
            round(
                float(
                    goal.get(
                        "current_saved",
                        0
                    )
                ),
                2
            ),
            goal.get(
                "deadline",
                ""
            ),
            goal.get(
                "priority",
                ""
            ),
            required_display,
            round(
                float(allocation),
                2
            ),
            goal.get(
                "status",
                ""
            )
        ])

    goal_df = pd.DataFrame(
        goal_rows,
        columns=[
            "Goal",
            "Target (AUD)",
            "Saved (AUD)",
            "Deadline",
            "Priority",
            "Minimum / Week",
            "Planned / Week",
            "Status"
        ]
    )

    # -----------------------------------------------------
    # SUGGESTIONS
    # -----------------------------------------------------

    suggestion_df = build_plan_suggestions(
        plan,
        expenses
    )

    gr.Info(
        "✅ Your plan has been calculated!"
    )

    return (
        state,
        income_card,
        expense_card,
        goal_card,
        status_card,
        status_message,
        summary_df,
        goal_df,
        suggestion_df
    )


# =========================================================
# GOAL MANAGEMENT
# =========================================================

def build_plan_display_from_state(state):

    plan = replan_from_state(state)

    if plan["status"] == "achievable":
        status_text = "✅ Achievable"

    elif plan["status"] == "needs attention":
        status_text = "⚠️ Needs attention"

    else:
        status_text = "⚠️ Not achievable"

    income_card = f"""
### 💰 Weekly Income
**${plan['general_income']:.2f} AUD**
"""

    expense_card = f"""
### 🛍️ Weekly Expenses
**${plan['total_expenses']:.2f} AUD**
"""

    goal_card = f"""
### 🎯 Available for Goals
**${plan['money_for_goals']:.2f} AUD**
"""

    status_card = f"""
### 💗 Plan Status
**{status_text}**
"""

    user_name = state.get(
        "profile",
        {}
    ).get(
        "name",
        ""
    )

    if plan["status"] == "achievable":

        status_message = f"""
### ✅ {user_name}, your plan is achievable

Your current weekly budget can cover the minimum
requirements for your active saving goals.
"""

    else:

        status_message = f"""
### ⚠️ {user_name}, your plan needs adjustment

Current weekly shortfall:
**${plan['shortfall']:.2f} AUD**

Check the Suggestions table below.
"""

    summary_df = pd.DataFrame(
        [
            [
                "Weekly Income",
                f"${plan['general_income']:.2f} AUD"
            ],
            [
                "Weekly Expenses",
                f"${plan['total_expenses']:.2f} AUD"
            ],
            [
                "Safety Buffer",
                f"${plan['buffer']:.2f} AUD"
            ],
            [
                "Available for Goals",
                f"${plan['money_for_goals']:.2f} AUD"
            ],
            [
                "Minimum Needed for Goals",
                f"${plan['total_required']:.2f} AUD"
            ],
            [
                "Plan Status",
                status_text
            ]
        ],
        columns=[
            "Item",
            "Weekly Amount / Result"
        ]
    )

    goal_rows = []

    for goal in plan["goals"]:

        goal_name = goal.get(
            "name",
            ""
        )

        required = goal.get(
            "required_per_period"
        )

        if required is not None:
            required = round(
                float(required),
                2
            )

        allocation = plan[
            "allocations"
        ].get(
            goal_name,
            0
        )

        goal_rows.append([
            goal_name,
            round(
                float(
                    goal.get(
                        "target_amount",
                        0
                    )
                ),
                2
            ),
            round(
                float(
                    goal.get(
                        "current_saved",
                        0
                    )
                ),
                2
            ),
            goal.get(
                "deadline",
                ""
            ),
            goal.get(
                "priority",
                ""
            ),
            required,
            round(
                float(allocation),
                2
            ),
            goal.get(
                "status",
                ""
            )
        ])

    goal_df = pd.DataFrame(
        goal_rows,
        columns=[
            "Goal",
            "Target (AUD)",
            "Saved (AUD)",
            "Deadline",
            "Priority",
            "Minimum / Week",
            "Planned / Week",
            "Status"
        ]
    )

    suggestion_df = build_plan_suggestions(
        plan,
        state.get(
            "expenses",
            []
        )
    )

    return (
        plan,
        income_card,
        expense_card,
        goal_card,
        status_card,
        status_message,
        summary_df,
        goal_df,
        suggestion_df
    )


def ui_load_managed_goal(
    state,
    goal_name
):

    if not state or not goal_name:
        return (
            "",
            None,
            "",
            None,
            None
        )

    for goal in state.get(
        "goals",
        []
    ):

        if (
            str(
                goal.get(
                    "name",
                    ""
                )
            ).lower()
            == str(goal_name).lower()
        ):

            return (
                goal.get(
                    "name",
                    ""
                ),
                goal.get(
                    "target_amount"
                ),
                goal.get(
                    "deadline",
                    ""
                ),
                goal.get(
                    "priority"
                ),
                goal.get(
                    "current_saved",
                    0
                )
            )

    return (
        "",
        None,
        "",
        None,
        None
    )


def ui_manage_goal(
    state,
    selected_goal,
    action,
    goal_name,
    target_amount,
    deadline,
    priority,
    current_saved
):

    if not state:
        raise gr.Error(
            "Please create your plan first."
        )

    if not action:
        raise gr.Error(
            "Please choose an action."
        )

    updated_state = copy.deepcopy(
        state
    )

    goals = updated_state.get(
        "goals",
        []
    )

    message = ""

    # -----------------------------------------------------
    # ADD
    # -----------------------------------------------------

    if action == "Add":

        if not goal_name or not goal_name.strip():
            raise gr.Error(
                "Please enter the new goal name."
            )

        if (
            target_amount is None
            or target_amount <= 0
        ):
            raise gr.Error(
                "Please enter a target amount greater than 0."
            )

        if not deadline or not deadline.strip():
            raise gr.Error(
                "Please enter a deadline."
            )

        try:
            datetime.strptime(
                deadline.strip(),
                "%Y-%m-%d"
            )

        except ValueError:
            raise gr.Error(
                "Deadline must use YYYY-MM-DD."
            )

        if not priority:
            raise gr.Error(
                "Please choose a priority."
            )

        if current_saved is None:
            current_saved = 0

        goals = add_goal(
            goals,
            goal_name.strip(),
            float(target_amount),
            deadline.strip(),
            priority,
            float(current_saved)
        )

        message = (
            f"### ✅ Goal added\n\n"
            f"**{goal_name.strip()}** has been added."
        )

    # -----------------------------------------------------
    # EDIT
    # -----------------------------------------------------

    elif action == "Edit":

        if not selected_goal:
            raise gr.Error(
                "Please choose an existing goal."
            )

        clean_deadline = None

        if deadline and deadline.strip():

            try:
                datetime.strptime(
                    deadline.strip(),
                    "%Y-%m-%d"
                )

            except ValueError:
                raise gr.Error(
                    "Deadline must use YYYY-MM-DD."
                )

            clean_deadline = deadline.strip()

        goals = edit_goal(
            goals,
            selected_goal,
            target_amount=(
                float(target_amount)
                if target_amount is not None
                else None
            ),
            deadline=clean_deadline,
            priority=priority
        )

        if current_saved is not None:

            if current_saved < 0:
                raise gr.Error(
                    "Current savings cannot be negative."
                )

            for goal in goals:

                if (
                    str(
                        goal.get(
                            "name",
                            ""
                        )
                    ).lower()
                    == str(
                        selected_goal
                    ).lower()
                ):
                    goal[
                        "current_saved"
                    ] = float(
                        current_saved
                    )

        message = (
            f"### ✅ Goal updated\n\n"
            f"**{selected_goal}** has been edited."
        )

    # -----------------------------------------------------
    # PAUSE
    # -----------------------------------------------------

    elif action == "Pause":

        if not selected_goal:
            raise gr.Error(
                "Please choose an existing goal."
            )

        goals = change_goal_status(
            goals,
            selected_goal,
            "paused"
        )

        message = (
            f"### ⏸️ Goal paused\n\n"
            f"**{selected_goal}** is temporarily paused."
        )

    # -----------------------------------------------------
    # RESUME
    # -----------------------------------------------------

    elif action == "Resume":

        if not selected_goal:
            raise gr.Error(
                "Please choose an existing goal."
            )

        goals = change_goal_status(
            goals,
            selected_goal,
            "active"
        )

        message = (
            f"### ▶️ Goal resumed\n\n"
            f"**{selected_goal}** is active again."
        )

    # -----------------------------------------------------
    # CANCEL
    # -----------------------------------------------------

    elif action == "Cancel":

        if not selected_goal:
            raise gr.Error(
                "Please choose an existing goal."
            )

        goals, released_money = cancel_goal(
            goals,
            selected_goal
        )

        message = (
            f"### ❌ Goal cancelled\n\n"
            f"**{selected_goal}** has been cancelled.\n\n"
            f"${released_money:.2f} AUD is no longer "
            "assigned to that goal."
        )

    updated_state[
        "goals"
    ] = goals

    (
        plan,
        income_card,
        expense_card,
        goal_card,
        status_card,
        status_message,
        summary_df,
        goal_df,
        suggestion_df
    ) = build_plan_display_from_state(
        updated_state
    )

    goal_names = [
        goal.get(
            "name"
        )
        for goal in goals
        if goal.get(
            "status"
        ) != "cancelled"
    ]

    gr.Info(
        "✅ Goal changes saved and plan recalculated."
    )

    return (
        updated_state,
        message,
        income_card,
        expense_card,
        goal_card,
        status_card,
        status_message,
        summary_df,
        goal_df,
        suggestion_df,
        gr.update(
            choices=goal_names,
            value=(
                goal_names[0]
                if goal_names
                else None
            )
        )
    )


def ui_recalculate_plan(
    state
):

    if not state:
        raise gr.Error(
            "Please create your plan first."
        )

    (
        plan,
        income_card,
        expense_card,
        goal_card,
        status_card,
        status_message,
        summary_df,
        goal_df,
        suggestion_df
    ) = build_plan_display_from_state(
        state
    )

    goal_names = [
        goal.get(
            "name"
        )
        for goal in state.get(
            "goals",
            []
        )
        if goal.get(
            "status"
        ) != "cancelled"
    ]

    gr.Info(
        "✅ Plan recalculated."
    )

    return (
        income_card,
        expense_card,
        goal_card,
        status_card,
        status_message,
        summary_df,
        goal_df,
        suggestion_df,
        gr.update(
            choices=goal_names,
            value=(
                goal_names[0]
                if goal_names
                else None
            )
        )
    )


# =========================================================
# WEEKLY TRACKING
# =========================================================

def normalise_transaction_date(
    date_value
):

    if not date_value:
        raise gr.Error(
            "Please enter a transaction date."
        )

    date_text = str(
        date_value
    ).strip()

    accepted_formats = [
        "%Y-%m-%d",
        "%Y/%m/%d"
    ]

    for date_format in accepted_formats:

        try:

            parsed = datetime.strptime(
                date_text,
                date_format
            )

            return parsed.strftime(
                "%Y-%m-%d"
            )

        except ValueError:
            pass

    raise gr.Error(
        "Please use YYYY-MM-DD or YYYY/MM/DD."
    )


def make_transaction_table(
    state
):

    rows = []

    for transaction in state.get(
        "transactions",
        []
    ):

        rows.append([
            transaction.get(
                "date",
                ""
            ),
            transaction.get(
                "category",
                ""
            ),
            round(
                float(
                    transaction.get(
                        "amount",
                        0
                    )
                ),
                2
            ),
            transaction.get(
                "description",
                ""
            )
        ])

    return pd.DataFrame(
        rows,
        columns=[
            "Date",
            "Category",
            "Amount (AUD)",
            "Description"
        ]
    )


def make_spending_comparison(
    state
):

    transactions = state.get(
        "transactions",
        []
    )

    expenses = state.get(
        "expenses",
        []
    )

    weekly_transactions = (
        filter_transactions_by_week(
            transactions
        )
    )

    comparison = compare_actual_to_plan(
        weekly_transactions,
        expenses,
        "weekly",
        period_complete=False
    )

    compared_categories = set()

    for row in comparison:

        compared_categories.add(
            normalise_expense_name(
                row.get(
                    "category",
                    ""
                )
            )
        )

    actual_totals = {}

    for transaction in weekly_transactions:

        category = normalise_expense_name(
            transaction.get(
                "category",
                ""
            )
        )

        actual_totals[
            category
        ] = (
            actual_totals.get(
                category,
                0
            )
            + float(
                transaction.get(
                    "amount",
                    0
                )
            )
        )

    # Include expenses that were not in the plan
    for category, actual in actual_totals.items():

        if category not in compared_categories:

            comparison.append({
                "category": category,
                "planned": 0,
                "actual": actual,
                "difference": actual,
                "status": "unplanned expense"
            })

    rows = []

    for row in comparison:

        rows.append([
            row.get(
                "category",
                ""
            ),
            round(
                float(
                    row.get(
                        "planned",
                        0
                    )
                ),
                2
            ),
            round(
                float(
                    row.get(
                        "actual",
                        0
                    )
                ),
                2
            ),
            round(
                float(
                    row.get(
                        "difference",
                        0
                    )
                ),
                2
            ),
            row.get(
                "status",
                ""
            )
        ])

    comparison_df = pd.DataFrame(
        rows,
        columns=[
            "Category",
            "Planned (AUD)",
            "Actual (AUD)",
            "Difference (AUD)",
            "Status"
        ]
    )

    return (
        comparison,
        comparison_df
    )

def make_spending_chart(state):

    comparison, _ = make_spending_comparison(
        state
    )

    if not comparison:
        return None

    categories = [
        str(row.get("category", "")).title()
        for row in comparison
    ]

    planned = [
        float(row.get("planned", 0))
        for row in comparison
    ]

    actual = [
        float(row.get("actual", 0))
        for row in comparison
    ]

    x = list(range(len(categories)))
    width = 0.35

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    ax.bar(
        [i - width / 2 for i in x],
        planned,
        width,
        label="Planned"
    )

    ax.bar(
        [i + width / 2 for i in x],
        actual,
        width,
        label="Actual"
    )

    ax.set_title(
        "Planned vs Actual Weekly Spending"
    )

    ax.set_xlabel(
        "Expense Category"
    )

    ax.set_ylabel(
        "Amount (AUD)"
    )

    ax.set_xticks(x)

    ax.set_xticklabels(
        categories,
        rotation=30,
        ha="right"
    )

    ax.legend()

    fig.tight_layout()

    return fig

def ui_add_transaction(
    state,
    date_value,
    category,
    amount,
    description
):

    if not state:
        raise gr.Error(
            "Please create your plan first."
        )

    if not category or not category.strip():
        raise gr.Error(
            "Please enter a transaction category."
        )

    if amount is None:
        raise gr.Error(
            "Please enter the transaction amount."
        )

    if amount < 0:
        raise gr.Error(
            "Transaction amount cannot be negative."
        )

    clean_date = normalise_transaction_date(
        date_value
    )

    updated_state = copy.deepcopy(
        state
    )

    transactions = updated_state.setdefault(
        "transactions",
        []
    )

    add_transaction(
        transactions,
        clean_date,
        category.strip(),
        float(amount),
        (
            description.strip()
            if description
            else ""
        )
    )

    transaction_df = (
        make_transaction_table(
            updated_state
        )
    )

    _, comparison_df = (
        make_spending_comparison(
            updated_state
        )
    )

    chart = make_spending_chart(
    updated_state
    )

    status = (
        "### ✅ Transaction added\n\n"
        f"**{category.strip()}** — "
        f"${float(amount):.2f} AUD "
        f"on {clean_date}"
    )

    gr.Info(
        "✅ Transaction added successfully."
    )

    return (
        updated_state,
        transaction_df,
        comparison_df,
        chart,
        status
    )


def ui_load_transactions_csv(
    state,
    uploaded_file
):

    if not state:
        raise gr.Error(
            "Please create your plan first."
        )

    if uploaded_file is None:
        raise gr.Error(
            "Please choose a CSV file first."
        )

    if isinstance(
        uploaded_file,
        str
    ):
        file_path = uploaded_file

    elif hasattr(
        uploaded_file,
        "name"
    ):
        file_path = uploaded_file.name

    else:
        file_path = str(
            uploaded_file
        )

    loaded = load_transactions_csv(
        file_path
    )

    valid_transactions = []

    skipped = 0

    for transaction in loaded:

        if float(
            transaction.get(
                "amount",
                0
            )
        ) < 0:

            skipped += 1
            continue

        valid_transactions.append(
            transaction
        )

    updated_state = copy.deepcopy(
        state
    )

    updated_state.setdefault(
        "transactions",
        []
    ).extend(
        valid_transactions
    )

    transaction_df = (
        make_transaction_table(
            updated_state
        )
    )

    _, comparison_df = (
        make_spending_comparison(
            updated_state
        )
    )

    chart = make_spending_chart(
    updated_state
    )

    status = (
        "### ✅ CSV loaded\n\n"
        f"Added **{len(valid_transactions)}** transactions."
    )

    if skipped > 0:

        status += (
            f"\n\nSkipped **{skipped}** "
            "negative transaction(s)."
        )

    gr.Info(
        "✅ CSV transactions loaded."
    )

    return (
        updated_state,
        transaction_df,
        comparison_df,
        chart,
        status
    )


def ui_generate_weekly_summary(
    state
):

    if not state:
        raise gr.Error(
            "Please create your plan first."
        )

    comparison, comparison_df = (
        make_spending_comparison(
            state
        )
    )

    plan = replan_from_state(
        state
    )

    impact = calculate_overspending_impact(
        comparison,
        plan[
            "money_for_goals"
        ],
        plan[
            "goals"
        ]
    )

    weekly_transactions = (
        filter_transactions_by_week(
            state.get(
                "transactions",
                []
            )
        )
    )

    planned_total = sum(
        float(
            row.get(
                "planned",
                0
            )
        )
        for row in comparison
    )

    actual_total = sum(
        float(
            transaction.get(
                "amount",
                0
            )
        )
        for transaction in weekly_transactions
    )

    difference = (
        actual_total
        - planned_total
    )

    summary_df = pd.DataFrame(
        [
            [
                "Transactions recorded",
                str(
                    len(
                        weekly_transactions
                    )
                )
            ],
            [
                "Planned spending",
                f"${planned_total:.2f} AUD"
            ],
            [
                "Actual spending so far",
                f"${actual_total:.2f} AUD"
            ],
            [
                "Difference",
                f"${difference:.2f} AUD"
            ],
            [
                "Overspending",
                f"${impact['total_overspending']:.2f} AUD"
            ],
            [
                "Goal money before overspending",
                (
                    f"${plan['money_for_goals']:.2f} AUD"
                )
            ],
            [
                "Goal money after overspending",
                (
                    f"${impact['money_for_goals_after_overspending']:.2f} AUD"
                )
            ],
            [
                "Goal status after tracking",
                impact[
                    "status"
                ]
            ]
        ],
        columns=[
            "Metric",
            "Result"
        ]
    )

    feedback = create_weekly_feedback(
        comparison,
        impact
    )

    advice = generate_weekly_advice(
        feedback
    )

    advice_text = (
        "### 💡 Weekly Advice\n\n"
        + advice
    )

    chart = make_spending_chart(
    state
    )

    return (
        comparison_df,
        chart,
        summary_df,
        advice_text
    )


# =========================================================
# AI FINANCE ASSISTANT
# =========================================================

def ui_finance_chat(
    state,
    history,
    question
):

    if not question or not question.strip():
        raise gr.Error(
            "Please enter a question."
        )

    if not state:
        raise gr.Error(
            "Please create your plan first."
        )

    plan = replan_from_state(
        state
    )

    expense_lines = []

    for expense in state.get(
        "expenses",
        []
    ):

        expense_lines.append(
            (
                f"- {expense.get('name')}: "
                f"${expense.get('amount')} "
                f"{expense.get('frequency')}, "
                f"{expense.get('classification')}"
            )
        )

    goal_lines = []

    for goal in plan.get(
        "goals",
        []
    ):

        required = goal.get(
            "required_per_period"
        )

        if required is None:
            required_text = "not currently calculable"

        else:
            required_text = (
                f"${required:.2f}/week"
            )

        goal_lines.append(
            (
                f"- {goal.get('name')}: "
                f"target ${goal.get('target_amount')}, "
                f"saved ${goal.get('current_saved')}, "
                f"deadline {goal.get('deadline')}, "
                f"priority {goal.get('priority')}, "
                f"minimum {required_text}"
            )
        )

    finance_context = f"""
USER FINANCIAL DATA

Weekly general income:
${plan['general_income']:.2f}

Weekly expenses:
${plan['total_expenses']:.2f}

Weekly safety buffer:
${plan['buffer']:.2f}

Money available for goals:
${plan['money_for_goals']:.2f}

Minimum required for goals:
${plan['total_required']:.2f}

Current plan status:
{plan['status']}

Expenses:
{chr(10).join(expense_lines)}

Saving goals:
{chr(10).join(goal_lines)}
"""

    prompt = f"""
Use the verified financial data below to answer
the user's finance question.

Do not invent amounts.

Do not modify the saved plan.

If the user asks a what-if question, explain it as
a hypothetical change using the supplied data.

Keep the answer simple and practical.

{finance_context}

USER QUESTION:
{question}
"""

    reply = ask_gemini(
        prompt
    )

    if history is None:
        history = []

    updated_history = list(
        history
    )

    updated_history.append({
        "role": "user",
        "content": question.strip()
    })

    updated_history.append({
        "role": "assistant",
        "content": reply
    })

    return (
        updated_history,
        ""
    )


# =========================================================
# SAVE / LOAD
# =========================================================

def ui_save_progress(
    state
):

    if not state:
        raise gr.Error(
            "Please create your plan first."
        )

    file_path = os.path.join(
        os.getcwd(),
        "finance_progress.json"
    )

    save_app_state(
        state,
        file_path
    )

    status = """
### ✅ Progress saved

Use the file above to download your current planner data.

You can upload it again next time to continue.
"""

    gr.Info(
        "✅ Progress file created."
    )

    return (
        file_path,
        status
    )


def ui_load_progress(
    uploaded_file
):

    if uploaded_file is None:
        raise gr.Error(
            "Please upload a saved progress file."
        )

    if isinstance(
        uploaded_file,
        str
    ):
        file_path = uploaded_file

    elif hasattr(
        uploaded_file,
        "name"
    ):
        file_path = uploaded_file.name

    else:
        file_path = str(
            uploaded_file
        )

    try:
        state = load_app_state(
            file_path
        )

    except Exception:
        raise gr.Error(
            "The progress file could not be loaded."
        )

    if not isinstance(
        state,
        dict
    ):
        raise gr.Error(
            "This is not a valid progress file."
        )

    (
        plan,
        income_card,
        expense_card,
        goal_card,
        status_card,
        status_message,
        summary_df,
        goal_df,
        suggestion_df
    ) = build_plan_display_from_state(
        state
    )

    income_df = pd.DataFrame(
        state.get(
            "income",
            []
        )
    )

    expense_df = pd.DataFrame(
        state.get(
            "expenses",
            []
        )
    )

    goal_input_df = pd.DataFrame(
        state.get(
            "goals",
            []
        )
    )

    transaction_df = (
        make_transaction_table(
            state
        )
    )

    _, comparison_df = (
        make_spending_comparison(
            state
        )
    )

    goal_names = [
        goal.get(
            "name"
        )
        for goal in state.get(
            "goals",
            []
        )
        if goal.get(
            "status"
        ) != "cancelled"
    ]

    status = """
### ✅ Progress loaded

Your saved financial plan has been restored.
"""

    gr.Info(
        "✅ Saved progress loaded."
    )

    return (
        state,
        state.get(
            "profile",
            {}
        ),
        income_df,
        expense_df,
        goal_input_df,
        income_card,
        expense_card,
        goal_card,
        status_card,
        status_message,
        summary_df,
        goal_df,
        suggestion_df,
        transaction_df,
        comparison_df,
        gr.update(
            choices=goal_names,
            value=(
                goal_names[0]
                if goal_names
                else None
            )
        ),
        status
    )
````
