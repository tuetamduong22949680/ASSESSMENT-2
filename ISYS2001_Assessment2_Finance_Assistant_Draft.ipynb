#ISYS2001 Assessment 2 – Finance Assistant

# Step 1 – Understand the Problem

## Problem

Managing personal finances can become difficult when a person has different income sources, many expenses, household responsibilities and multiple saving goals.

A person's financial situation may also change over time.

Income may be regular or variable. Expenses may be fixed, flexible, seasonal, temporary or one-off. Some costs may be shared with other people, while some income may only be available for a specific person or purpose.

Because of this, a simple fixed budget may not accurately represent the user's real financial situation.

Users may know roughly how much they earn and spend, but they may not know:

- how much of their income is actually available to use
- how much of each shared expense they are personally responsible for
- how much money is left after essential expenses
- how much money should be kept as a safety or emergency buffer
- how much they can realistically save
- whether they can reach multiple saving goals by their deadlines
- which goals should be prioritised when money is limited
- which expenses could realistically be reduced
- how changes in income or spending affect their saving goals
- whether they have forgotten important expenses when creating their budget

## Proposed Solution

The project will create a **Smart Savings and Spending Planner**.

The application will build a personalised financial plan using information about the user's:

- personal and household profile
- income sources
- financial responsibilities
- expenses
- existing savings
- saving goals

The profile will help the application understand the user's situation and decide which possible expenses and follow-up questions are relevant.

## Profile

The user can build their profile by selecting different options that apply to their situation.

The profile can include:

- study or employment situation
- people living with the user
- people the user supports outside the household
- dependants
- shared responsibilities
- pets
- other relevant household situations

The profile will be flexible so that multiple situations can be combined instead of forcing every user into one fixed family type.

Uncommon situations can be entered using an **Other** option and Gemini can help interpret the description.

## Income

The application will support multiple income sources.

Income information may include:

- amount
- payment frequency
- person receiving the income
- whether the income is regular, variable, seasonal, temporary or one-off
- how much of the income is actually available for the budget
- whether the income is restricted to a specific purpose

For variable income, the application can track income across multiple payment periods instead of assuming that one period represents the user's normal income.

The user will enter take-home or usable income where possible so that the budget is based on money that can actually be spent.

## Expenses

The user will first describe their expenses naturally instead of completing a very long checklist.

Gemini will extract as much information as possible from the user's description, including:

- expense category
- amount
- frequency
- who the expense relates to
- who pays the expense
- whether the expense is shared
- whether the expense is regular, variable, seasonal, temporary or one-off

The extracted information will be shown back to the user for confirmation before calculations are performed.

The application will then compare the extracted expenses with a list of possible expenses based on the user's profile.

Expenses already identified will be filled automatically.

Only possible missing expenses will be shown to the user.

For example, if a user enters rent but does not mention household bills, the application may ask whether electricity, water, gas or internet are included in the rent.

Missing information will not automatically be treated as zero.

It can instead be recorded as:

- unknown
- not applicable
- paid by someone else
- not yet entered

Expenses will also be classified into useful budgeting groups such as:

- fixed expenses
- essential flexible expenses
- optional flexible expenses

The user will be able to correct the classification if necessary.

Refunds, reimbursements and transfers between the user's own accounts should not automatically be treated as normal income or normal spending.

## Saving Goals

The application will support multiple saving goals.

Each goal may contain:

- goal name
- target amount
- currency
- amount already saved for that goal
- deadline
- priority
- whether the deadline can be changed
- the user's share of the goal if another person is contributing

The user may also have existing savings that have not yet been assigned to a specific goal.

These will be kept separate so that the same savings are not accidentally counted towards multiple goals.

Goals may also be ongoing or have no fixed deadline, such as building an emergency fund.

## Financial Calculations

Python will perform the main financial calculations.

The calculations will include:

- normalising different income and expense frequencies
- calculating usable income
- calculating the user's actual share of expenses
- calculating total spending
- calculating available money after expenses
- allowing for an emergency or safety buffer where appropriate
- calculating how much must be saved for each goal
- checking whether goals can be reached by their deadlines
- comparing multiple goals using their priorities and deadlines

Gemini will support the conversation and explain results, but the main financial numbers will come from Python calculations using the user's data.

## Building the Plan

If the user's current income and spending are enough to reach the goals, the application will create a suggested saving plan.

If the goals are not achievable, the application will look for realistic alternatives.

Possible suggestions may include:

- reducing optional flexible expenses
- reducing some essential flexible expenses where reasonable
- using unallocated savings
- changing a goal amount
- extending a flexible deadline
- changing the priority of a lower-priority goal

The application should avoid suggesting reductions to fixed or essential expenses before more flexible options are considered.

The user can accept or reject each suggested change.

The application will recalculate the plan after each accepted change.

If the user rejects the available changes and the goals still cannot be reached, the application will explain that it cannot find a realistic way to achieve all of the goals under the current conditions.

It may then allow the user to change a deadline, goal amount or priority, or continue with the current plan while showing the expected shortfall.

## Ongoing Tracking

After a plan is created, the user can continue tracking actual income and spending.

Income can be recorded each payment cycle and expenses can be entered manually or loaded from transaction data where appropriate.

The application can compare:

- expected income against actual income
- planned spending against actual spending
- planned saving against actual saving
- expected goal progress against actual progress

If the user's financial situation changes, the application can recalculate the plan.

For example, if income decreases or spending increases, the application can show how this may affect the expected completion date of a saving goal.

It is not intended to provide professional investment, tax or legal advice.

## Main Goal

The main goal of the application is to help users create a realistic and personalised saving and spending plan based on their actual financial situation.

The application will combine Gemini for conversation and understanding natural-language input with Python for financial calculations, planning and ongoing tracking.

# 24. Important Rule for Monthly and Annual Bills

A monthly or annual bill needs special treatment when the user uses a weekly budget.

For example:

Phone bill:

**$60 per month**

Weekly budget provision:

**approximately $13.85 per week**

The $13.85 represents money that should be set aside over time for the future phone bill.

When the actual $60 payment happens, the application must not treat:

- the weekly provisions, and
- the $60 payment

as two separate expenses.

That would double count the same cost.

The same rule applies to:

- annual insurance
- registration
- school fees
- quarterly utilities
- other less-frequent bills

The application should distinguish between:

1. money being reserved for a future expense, and
2. the actual payment of that expense.

If a large bill is due soon and the user has not already saved enough for it, simply dividing the annual or monthly cost into future weekly amounts may not be enough.

The application may need to calculate a temporary catch-up amount based on:

- amount still required
- money already reserved
- time until the due date

This prevents a bill from appearing affordable on average when the user does not actually have enough money available when it becomes due.

# 25. Important Rule for Restricted Money

Restricted money must remain separate until it is used for its intended purpose.

In this example:

Child support = $120 per week

Childcare = $220 per week

The $120 can reduce the amount of childcare that must be paid using general money.

If restricted income is greater than the related expense, the remaining restricted amount should not automatically become general saving money.

It should remain restricted unless the user confirms that it can legally or practically be used for another purpose.

---

# 26. Important Rule for Contributions From Other People

There are two different situations.

### Situation A – Person Transfers Money Into the Budget

Example:

Partner transfers $350 into the shared household account.

This can be recorded as a household contribution.

### Situation B – Person Pays an Expense Directly

Example:

Partner directly pays $300 of a $600 rent bill.

The user's expense responsibility is only $300.

The partner's $300 should not also be added as income.

The application must use one method or the other for the same money.

---

# 27. Important Rule for Current Expenses

All confirmed current expenses should be included when calculating the user's starting financial position.

This includes:

- fixed expenses
- essential flexible expenses
- optional flexible expenses

The categories are used to decide which expenses may reasonably be changed.

They are not used to remove optional spending from the starting budget.

For example:

If the user currently spends $80 per week going out, that $80 must first appear in the current budget.

Only after the plan is found to be unachievable should the application suggest reducing it.

---

# 28. Important Rule for Existing Savings

Existing savings are a balance, not income.

Example:

The user already has:

- $300 for the Flight
- $200 for the PC

The $500 should reduce the remaining amount required for those goals.

It should not also be added to weekly income.

If the user has general savings that are not assigned to a goal, they should be stored as:

**Unallocated savings**

The user can later choose whether to use them.

---

# 29. Important Rule for Goal Contributions

Money contributed by another person directly toward a goal should reduce the user's required share.

Example:

Trip target:

$4,000

Partner will contribute:

$2,000

User responsibility:

$2,000

The partner's $2,000 should not also be added to the user's general income.

---

# 30. Important Rule for Transfers

Moving money between the user's own accounts does not create income or spending.

Example:

$200 transferred from everyday account to savings account.

This is:

**Transfer / allocation**

It is not:

**$200 expense**

Later moving the same $200 back into the everyday account is also not new income.

---

# 31. Important Rule for Credit Cards

A purchase should only be counted once.

Example:

The user spends $100 on groceries using a credit card.

Expense:

**Groceries = $100**

Later the user transfers $100 from their bank account to pay the credit card.

That payment should not create another $100 grocery expense.

However:

- credit card interest
- late fees
- other fees

may be separate real expenses.

---

# 32. Important Rule for Missing Information

The application must distinguish between:

- $0
- Unknown
- Missing
- Not applicable
- Paid by another person

These values have different meanings.

Example:

Electricity = $0

means the user confirms there is no electricity cost.

Electricity = Missing

means the application does not yet know the amount.

The program must not silently replace missing values with zero.

---

# 33. Important Rule for Variable Income

A missing income record should not be interpreted as zero income.

For example:

Week 1 = $500

Week 2 = $0

Week 3 = Missing

means:

- Week 2 is a confirmed zero-income week
- Week 3 has no information yet

Only confirmed values should be included according to the selected averaging method.

---

# 34. Important Rule for Time

Financial calculations must consider timing, not only averages.

Two plans with the same yearly total may still be different if:

- a large bill is due next week
- a saving goal is due next month
- income ends in three months
- an expense only exists for part of the year

Therefore, when timing matters, the application should consider:

- start date
- end date
- due date
- goal deadline
- number of remaining budgeting periods

---

# 35. Important Rule for Currency

Amounts in different currencies cannot be added directly.

If the base currency is AUD, a goal in another currency must first be converted into AUD before feasibility is calculated.

The original amount and currency should still be kept for display.

Example:

Original goal:

14,000,000 VND

Converted budgeting value:

AUD equivalent using the selected exchange rate.

The conversion should only happen once for each calculation.

# 3. Prevent Income Double Counting

FUNCTION validate_income_sources

    FOR each income source

        CHECK whether the money is:
            actual income
            household contribution
            restricted contribution
            one-off money
            transfer between own accounts

        IF source is an internal transfer
            DO NOT count as income
        END IF

        IF another person directly pays an expense
            DO NOT also record the same amount as income contribution
        END IF

        IF contribution is made directly to a saving goal
            DO NOT also add the same contribution to general income
        END IF

    END FOR

    RETURN cleaned income list

END FUNCTION

## Step 5A – Core Financial Calculations

I will build the program in small functions instead of writing the whole application at once.

I will begin with frequency normalisation so that income and expenses with different payment frequencies can be compared using the same budgeting period.

```python
def normalise_frequency(amount, frequency, budget_period="weekly"):
    """
    Convert an amount into the selected budgeting period.
    """

    if amount < 0:
        raise ValueError("Amount cannot be negative.")

    frequency = frequency.lower()
    budget_period = budget_period.lower()

    # Convert the original amount to a yearly amount first
    if frequency == "weekly":
        yearly_amount = amount * 52

    elif frequency == "fortnightly":
        yearly_amount = amount * 26

    elif frequency == "monthly":
        yearly_amount = amount * 12

    elif frequency == "yearly":
        yearly_amount = amount

    else:
        raise ValueError("Frequency must be weekly, fortnightly, monthly or yearly.")

    # Convert the yearly amount into the selected budgeting period
    if budget_period == "weekly":
        return yearly_amount / 52

    elif budget_period == "fortnightly":
        return yearly_amount / 26

    elif budget_period == "monthly":
        return yearly_amount / 12

    else:
        raise ValueError("Budget period must be weekly, fortnightly or monthly.")
```

```python
print(normalise_frequency(60, "monthly", "weekly"))
print(normalise_frequency(1000, "weekly", "monthly"))
print(normalise_frequency(1200, "yearly", "monthly"))
```

```python
def calculate_average_income(income_records):
    """
    Calculate the average income using confirmed records only.
    Missing records are ignored.
    """

    confirmed_amounts = []

    for record in income_records:
        if record["status"] == "entered":
            confirmed_amounts.append(record["amount"])

        elif record["status"] == "zero":
            confirmed_amounts.append(0)

        elif record["status"] == "missing":
            continue

    if len(confirmed_amounts) == 0:
        raise ValueError("No confirmed income records available.")

    return sum(confirmed_amounts) / len(confirmed_amounts)
```

```python
income_records = [
    {"amount": 450, "status": "entered"},
    {"amount": 300, "status": "entered"},
    {"amount": 0, "status": "zero"},
    {"amount": None, "status": "missing"}
]

print(calculate_average_income(income_records))
```

```python
def calculate_user_share(total_amount, share_percent=100):
    """
    Calculate the user's share of a shared expense.
    """

    if total_amount < 0:
        raise ValueError("Expense amount cannot be negative.")

    if share_percent < 0 or share_percent > 100:
        raise ValueError("Share percentage must be between 0 and 100.")

    user_share = total_amount * share_percent / 100

    return user_share
```

```python
print(calculate_user_share(600, 50))
print(calculate_user_share(600, 37))
print(calculate_user_share(300))
```

```python
def calculate_available_income(amount, availability_type="all", availability_value=None):
    """
    Calculate how much of an income source is available to the budget.
    """

    if amount < 0:
        raise ValueError("Income amount cannot be negative.")

    availability_type = availability_type.lower()

    if availability_type == "all":
        return amount

    elif availability_type == "percent":
        if availability_value is None:
            raise ValueError("Please provide an availability percentage.")

        if availability_value < 0 or availability_value > 100:
            raise ValueError("Percentage must be between 0 and 100.")

        return amount * availability_value / 100

    elif availability_type == "fixed":
        if availability_value is None:
            raise ValueError("Please provide a fixed available amount.")

        if availability_value < 0 or availability_value > amount:
            raise ValueError("Fixed available amount must be between 0 and the income amount.")

        return availability_value

    elif availability_type == "none":
        return 0

    else:
        raise ValueError("Availability type must be all, percent, fixed or none.")
```

```python
print(calculate_available_income(1000, "all"))
print(calculate_available_income(1000, "percent", 40))
print(calculate_available_income(1000, "fixed", 300))
print(calculate_available_income(1000, "none"))
```

```python
def calculate_income_totals(income_sources, budget_period="weekly"):
    """
    Calculate general and restricted usable income
    in the selected budgeting period.
    """

    general_income = 0
    restricted_income = 0

    for source in income_sources:

        # Work out how much of this income is actually available
        available_amount = calculate_available_income(
            source["amount"],
            source.get("availability_type", "all"),
            source.get("availability_value")
        )

        # Convert it into the selected budgeting period
        normalised_amount = normalise_frequency(
            available_amount,
            source["frequency"],
            budget_period
        )

        # Keep restricted money separate from general money
        if source.get("restricted", False):
            restricted_income += normalised_amount
        else:
            general_income += normalised_amount

    return general_income, restricted_income
```

```python
def apply_restricted_income(expense_amount, restricted_amount):
    """
    Apply restricted income to a related expense.

    Returns:
        amount_paid_by_restricted_income
        amount_left_for_general_budget
        restricted_income_left
    """

    if expense_amount < 0 or restricted_amount < 0:
        raise ValueError("Amounts cannot be negative.")

    amount_used = min(expense_amount, restricted_amount)

    general_budget_amount = expense_amount - amount_used

    restricted_left = restricted_amount - amount_used

    return amount_used, general_budget_amount, restricted_left
```

```python
def calculate_total_expenses(expenses, budget_period="weekly"):
    """
    Calculate the user's total confirmed expenses
    in the selected budgeting period.
    """

    total_expenses = 0

    for expense in expenses:

        # Ignore expenses that should not be counted
        if expense.get("status") in ["not_applicable", "paid_by_other", "missing", "unknown"]:
            continue

        # Calculate the user's share
        user_amount = calculate_user_share(
            expense["amount"],
            expense.get("share_percent", 100)
        )

        # Convert to selected budgeting period
        normalised_amount = normalise_frequency(
            user_amount,
            expense["frequency"],
            budget_period
        )

        total_expenses += normalised_amount

    return total_expenses
```

```python
def calculate_money_remaining(general_income, total_expenses):
    """
    Calculate how much general-purpose money remains
    after current expenses.
    """

    if general_income < 0:
        raise ValueError("General income cannot be negative.")

    if total_expenses < 0:
        raise ValueError("Total expenses cannot be negative.")

    return general_income - total_expenses
```

```python
def calculate_suggested_buffer(general_income, buffer_percent=5):
    """
    Suggest a safety buffer as a percentage of general usable income.
    """

    if general_income < 0:
        raise ValueError("General income cannot be negative.")

    if buffer_percent < 0 or buffer_percent > 100:
        raise ValueError("Buffer percentage must be between 0 and 100.")

    return general_income * buffer_percent / 100
```

```python
def calculate_money_for_goals(money_remaining, buffer_amount):
    """
    Calculate money available for saving goals
    after the confirmed safety buffer.
    """

    if buffer_amount < 0:
        raise ValueError("Buffer amount cannot be negative.")

    if buffer_amount > max(money_remaining, 0):
        raise ValueError("Safety buffer is larger than the money available.")

    return money_remaining - buffer_amount
```

```python
def calculate_goal_requirement(target_amount, already_saved, periods_left):
    """
    Calculate the remaining amount and required saving
    per budgeting period for a saving goal.
    """

    if target_amount < 0 or already_saved < 0:
        raise ValueError("Goal amounts cannot be negative.")

    remaining_amount = max(target_amount - already_saved, 0)

    # Goal is already completed
    if remaining_amount == 0:
        return 0, 0

    if periods_left <= 0:
        raise ValueError("Periods left must be greater than 0 for an incomplete goal.")

    required_per_period = remaining_amount / periods_left

    return remaining_amount, required_per_period
```

```python
def check_goal_feasibility(money_for_goals, required_amounts):
    """
    Check whether the available money is enough
    to cover all saving goal requirements.
    """

    if money_for_goals < 0:
        raise ValueError("Money available for goals cannot be negative.")

    total_required = sum(required_amounts)

    difference = money_for_goals - total_required

    if difference >= 0:
        status = "achievable"
        surplus = difference
        shortfall = 0
    else:
        status = "not achievable"
        surplus = 0
        shortfall = abs(difference)

    return total_required, status, surplus, shortfall
```

```python
def suggest_expense_reduction(current_amount, shortfall):
    """
    Suggest how much to reduce an optional expense
    to cover the current shortfall.
    """

    if current_amount < 0 or shortfall < 0:
        raise ValueError("Amounts cannot be negative.")

    reduction = min(current_amount, shortfall)

    suggested_amount = current_amount - reduction

    return reduction, suggested_amount
```

```python
def recalculate_after_reduction(
    general_income,
    total_expenses,
    buffer_amount,
    required_amounts,
    reduction_amount
):
    """
    Recalculate the budget after the user accepts
    an expense reduction.
    """

    if reduction_amount < 0:
        raise ValueError("Reduction amount cannot be negative.")

    if reduction_amount > total_expenses:
        raise ValueError("Reduction cannot be greater than total expenses.")

    new_total_expenses = total_expenses - reduction_amount

    new_money_remaining = calculate_money_remaining(
        general_income,
        new_total_expenses
    )

    new_money_for_goals = calculate_money_for_goals(
        new_money_remaining,
        buffer_amount
    )

    total_required, status, surplus, shortfall = check_goal_feasibility(
        new_money_for_goals,
        required_amounts
    )

    return (
        new_total_expenses,
        new_money_remaining,
        new_money_for_goals,
        status,
        surplus,
        shortfall
    )
```

```python
def sort_goals_by_priority(goals):
    """
    Sort active goals by:
    1. Priority
    2. Nearest deadline / fewest periods left
    """

    priority_order = {
        "must have": 1,
        "important": 2,
        "nice to have": 3
    }

    active_goals = []

    for goal in goals:
        if goal.get("status", "active") == "active":
            active_goals.append(goal)

    def goal_sort_key(goal):
        priority_rank = priority_order.get(
            goal["priority"].lower(),
            99
        )

        periods_left = goal.get("periods_left")

        # Goals with no deadline go after timed goals
        if periods_left is None:
            periods_left = 999999

        return priority_rank, periods_left

    return sorted(active_goals, key=goal_sort_key)
```

```python
def check_goal_completion(goal):
    """
    Check whether a saving goal has reached its target
    without changing paused or cancelled goals back to active.
    """

    target_amount = goal["target_amount"]
    current_saved = goal["current_saved"]

    if target_amount < 0 or current_saved < 0:
        raise ValueError("Goal amounts cannot be negative.")

    # Do not reactivate cancelled goals
    if goal.get("status") == "cancelled":
        return goal, f"{goal['name']} is cancelled."

    if current_saved >= target_amount:
        goal["status"] = "completed"

        message = (
            f"Congratulations! You reached your "
            f"{goal['name']} goal of ${target_amount:.2f}."
        )

    else:
        remaining = target_amount - current_saved

        # Keep paused goals paused
        if goal.get("status") != "paused":
            goal["status"] = "active"

        message = (
            f"{goal['name']} still needs "
            f"${remaining:.2f}."
        )

    return goal, message
```

```python
def update_goal_progress(goal, amount_saved):
    """
    Add new savings to a goal and check its status.
    """

    if amount_saved < 0:
        raise ValueError("Saved amount cannot be negative.")

    if goal.get("status") == "cancelled":
        raise ValueError("Cannot add savings to a cancelled goal.")

    if goal.get("status") == "completed":
        return goal, f"{goal['name']} is already completed."

    goal["current_saved"] += amount_saved

    return check_goal_completion(goal)
```

## Design Note – Parallel Goal Saving

Saving goals should normally run at the same time.

Each active goal receives the minimum amount needed per budgeting period to meet its deadline.

Priority does not mean that one goal must be completed before saving for another goal.

Priority is mainly used when:

- there is not enough money to fully fund every goal
- extra money is available and the user wants to complete a higher-priority goal earlier
- the application needs to suggest which lower-priority goal could be delayed or adjusted

If enough money is available, all active goals should receive their required saving amount during the same budgeting period.

```python
def allocate_goal_savings(goals, money_for_goals):
    """
    Allocate the required saving amount to all active goals
    when enough money is available.
    """

    if money_for_goals < 0:
        raise ValueError("Money available for goals cannot be negative.")

    active_goals = []

    for goal in goals:
        if goal.get("status", "active") == "active":
            active_goals.append(goal)

    total_required = 0

    for goal in active_goals:
        total_required += goal["required_per_period"]

    # Not enough money for all goals
    if money_for_goals < total_required:
        shortfall = total_required - money_for_goals

        return {
            "status": "not achievable",
            "allocations": {},
            "extra_money": 0,
            "shortfall": shortfall
        }

    # Enough money: fund all goals at the same time
    allocations = {}

    for goal in active_goals:
        allocations[goal["name"]] = goal["required_per_period"]

    extra_money = money_for_goals - total_required

    return {
        "status": "achievable",
        "allocations": allocations,
        "extra_money": extra_money,
        "shortfall": 0
    }
```

```python
def allocate_extra_to_priority_goals(goals, allocations, extra_money):
    """
    Add extra money to active goals by priority.

    Extra money is only allocated after every goal has
    already received its required amount for the period.
    """

    if extra_money < 0:
        raise ValueError("Extra money cannot be negative.")

    updated_allocations = allocations.copy()

    sorted_goals = sort_goals_by_priority(goals)

    for goal in sorted_goals:

        if extra_money <= 0:
            break

        current_allocation = updated_allocations.get(goal["name"], 0)

        remaining_amount = goal["remaining_amount"]

        # Do not allocate more than the goal still needs
        extra_needed = max(
            remaining_amount - current_allocation,
            0
        )

        extra_for_goal = min(extra_money, extra_needed)

        updated_allocations[goal["name"]] = (
            current_allocation + extra_for_goal
        )

        extra_money -= extra_for_goal

    return updated_allocations, extra_money
```

```python
def apply_goal_allocations(goals, allocations):
    """
    Apply this period's allocations to active goals.

    If an allocation is larger than the amount still needed,
    the unused money is returned instead of disappearing.
    """

    messages = []
    unused_money = 0

    for goal in goals:

        if goal.get("status", "active") != "active":
            continue

        amount = allocations.get(goal["name"], 0)

        if amount < 0:
            raise ValueError("Goal allocation cannot be negative.")

        remaining = max(
            goal["target_amount"] - goal["current_saved"],
            0
        )

        actual_amount = min(amount, remaining)

        # Money that was allocated but no longer needed
        unused_money += amount - actual_amount

        goal["current_saved"] += actual_amount

        updated_goal, message = check_goal_completion(goal)

        messages.append(message)

    return goals, messages, unused_money
```

```python
def reallocate_unused_money(goals, unused_money):
    """
    Move unused money to other active goals by priority.

    Money will not be added beyond a goal's target.
    Any money that still cannot be allocated is returned.
    """

    if unused_money < 0:
        raise ValueError("Unused money cannot be negative.")

    messages = []

    sorted_goals = sort_goals_by_priority(goals)

    for goal in sorted_goals:

        if unused_money <= 0:
            break

        remaining = max(
            goal["target_amount"] - goal["current_saved"],
            0
        )

        if remaining == 0:
            continue

        amount_to_add = min(unused_money, remaining)

        goal["current_saved"] += amount_to_add
        unused_money -= amount_to_add

        updated_goal, message = check_goal_completion(goal)

        messages.append(
            f"${amount_to_add:.2f} was moved to {goal['name']}. "
            + message
        )

    return goals, unused_money, messages
```

```python
def add_goal(
    goals,
    name,
    target_amount,
    current_saved=0,
    priority="Important",
    periods_left=None
):
    """
    Add a new saving goal while the plan is already running.
    """

    if target_amount <= 0:
        raise ValueError("Target amount must be greater than 0.")

    if current_saved < 0:
        raise ValueError("Current savings cannot be negative.")

    valid_priorities = [
        "must have",
        "important",
        "nice to have"
    ]

    if priority.lower() not in valid_priorities:
        raise ValueError(
            "Priority must be Must have, Important or Nice to have."
        )

    # Goal names must be unique
    for goal in goals:
        if goal["name"].lower() == name.lower():
            raise ValueError("A goal with this name already exists.")

    if current_saved >= target_amount:
        status = "completed"
    else:
        status = "active"

    new_goal = {
        "name": name,
        "target_amount": target_amount,
        "current_saved": current_saved,
        "priority": priority,
        "periods_left": periods_left,
        "status": status
    }

    goals.append(new_goal)

    return goals
```

```python
def change_goal_status(goals, goal_name, new_status):
    """
    Change a goal between active and paused.
    """

    valid_statuses = [
        "active",
        "paused"
    ]

    if new_status.lower() not in valid_statuses:
        raise ValueError(
            "Status must be active or paused."
        )

    for goal in goals:

        if goal["name"].lower() == goal_name.lower():

            if goal.get("status") == "completed":
                raise ValueError(
                    "A completed goal cannot be changed back."
                )

            if goal.get("status") == "cancelled":
                raise ValueError(
                    "A cancelled goal cannot be changed."
                )

            goal["status"] = new_status.lower()

            return goals

    raise ValueError("Goal not found.")
```

```python
def cancel_goal(goals, goal_name):
    """
    Cancel a goal and release the savings
    currently stored in that goal.
    """

    for goal in goals:

        if goal["name"].lower() == goal_name.lower():

            if goal.get("status") == "completed":
                raise ValueError(
                    "A completed goal should stay in history."
                )

            if goal.get("status") == "cancelled":
                raise ValueError(
                    "This goal is already cancelled."
                )

            released_money = goal["current_saved"]

            goal["current_saved"] = 0
            goal["status"] = "cancelled"

            return goals, released_money

    raise ValueError("Goal not found.")
```

```python
def edit_goal(
    goals,
    goal_name,
    new_target=None,
    new_priority=None,
    new_periods_left=None
):
    """
    Edit an active or paused goal.

    Returns:
        updated goals
        money released if target is reduced below current savings
    """

    valid_priorities = [
        "must have",
        "important",
        "nice to have"
    ]

    for goal in goals:

        if goal["name"].lower() == goal_name.lower():

            if goal.get("status") == "completed":
                raise ValueError(
                    "Completed goals should stay in history. "
                    "Create a new goal instead."
                )

            if goal.get("status") == "cancelled":
                raise ValueError("Cannot edit a cancelled goal.")

            released_money = 0

            # Change target
            if new_target is not None:

                if new_target <= 0:
                    raise ValueError(
                        "Target amount must be greater than 0."
                    )

                goal["target_amount"] = new_target

                # User has already saved more than the new target
                if goal["current_saved"] > new_target:

                    released_money = (
                        goal["current_saved"] - new_target
                    )

                    goal["current_saved"] = new_target
                    goal["status"] = "completed"

                elif goal["current_saved"] == new_target:
                    goal["status"] = "completed"

            # Change priority
            if new_priority is not None:

                if new_priority.lower() not in valid_priorities:
                    raise ValueError(
                        "Priority must be Must have, "
                        "Important or Nice to have."
                    )

                goal["priority"] = new_priority

            # Change deadline / periods left
            if new_periods_left is not None:

                if new_periods_left <= 0:
                    raise ValueError(
                        "Periods left must be greater than 0."
                    )

                goal["periods_left"] = new_periods_left

            return goals, released_money

    raise ValueError("Goal not found.")
```

```python
def update_goal_requirements(goals):
    """
    Recalculate remaining amount and required saving
    for each active goal.
    """

    for goal in goals:

        if goal.get("status") != "active":
            goal["remaining_amount"] = max(
                goal["target_amount"] - goal["current_saved"],
                0
            )
            goal["required_per_period"] = 0
            continue

        remaining_amount = max(
            goal["target_amount"] - goal["current_saved"],
            0
        )

        goal["remaining_amount"] = remaining_amount

        periods_left = goal.get("periods_left")

        if remaining_amount == 0:
            goal["required_per_period"] = 0
            goal["status"] = "completed"

        elif periods_left is None:
            goal["required_per_period"] = 0

        elif periods_left <= 0:
            goal["required_per_period"] = None

        else:
            goal["required_per_period"] = (
                remaining_amount / periods_left
            )

    return goals
```

```python
def check_updated_plan(goals, money_for_goals):
    """
    Recheck whether the current money available
    is enough for all active timed goals.
    """

    if money_for_goals < 0:
        raise ValueError("Money available for goals cannot be negative.")

    total_required = 0
    overdue_goals = []

    for goal in goals:

        if goal.get("status") != "active":
            continue

        required = goal.get("required_per_period")

        # None means the goal has money remaining
        # but no valid periods left
        if required is None:
            overdue_goals.append(goal["name"])
            continue

        total_required += required

    if overdue_goals:
        return {
            "status": "needs attention",
            "total_required": total_required,
            "surplus": 0,
            "shortfall": 0,
            "overdue_goals": overdue_goals
        }

    difference = money_for_goals - total_required

    if difference >= 0:
        status = "achievable"
        surplus = difference
        shortfall = 0
    else:
        status = "not achievable"
        surplus = 0
        shortfall = abs(difference)

    return {
        "status": status,
        "total_required": total_required,
        "surplus": surplus,
        "shortfall": shortfall,
        "overdue_goals": []
    }
```

```python
def find_adjustable_goal(goals):
    """
    Find the lowest-priority active goal that could be adjusted.
    Nice to have goals are considered before Important goals.
    Must have goals are not automatically suggested for adjustment.
    """

    priority_order = {
        "nice to have": 1,
        "important": 2,
        "must have": 3
    }

    adjustable_goals = []

    for goal in goals:

        if goal.get("status") != "active":
            continue

        if goal["priority"].lower() == "must have":
            continue

        adjustable_goals.append(goal)

    if len(adjustable_goals) == 0:
        return None

    adjustable_goals.sort(
        key=lambda goal: (
            priority_order.get(goal["priority"].lower(), 99),
            -goal.get("periods_left", 0)
        )
    )

    return adjustable_goals[0]
```

```python
import math

def suggest_deadline_extension(goal, shortfall):
    """
    Suggest a longer deadline for a flexible goal
    so its required saving per period becomes lower.
    """

    if shortfall < 0:
        raise ValueError("Shortfall cannot be negative.")

    current_required = goal["required_per_period"]
    remaining_amount = goal["remaining_amount"]

    if current_required is None or current_required <= 0:
        return None

    # How much this goal would need to contribute
    # after covering the current shortfall
    new_required = current_required - shortfall

    if new_required <= 0:
        return None

    new_periods = math.ceil(
        remaining_amount / new_required
    )

    suggested_required = (
        remaining_amount / new_periods
    )

    return new_periods, suggested_required
```

```python
def find_expense_reduction_options(
    expenses,
    shortfall,
    budget_period="weekly"
):
    """
    Find optional flexible expenses that could be reduced
    to help cover a saving-plan shortfall.
    """

    if shortfall < 0:
        raise ValueError("Shortfall cannot be negative.")

    options = []

    for expense in expenses:

        if expense.get("status") != "confirmed":
            continue

        if expense.get("classification") != "optional flexible":
            continue

        user_amount = calculate_user_share(
            expense["amount"],
            expense.get("share_percent", 100)
        )

        normalised_amount = normalise_frequency(
            user_amount,
            expense["frequency"],
            budget_period
        )

        possible_reduction = min(
            normalised_amount,
            shortfall
        )

        suggested_amount = (
            normalised_amount - possible_reduction
        )

        options.append({
            "name": expense["name"],
            "current_amount": normalised_amount,
            "possible_reduction": possible_reduction,
            "suggested_amount": suggested_amount
        })

    # Review larger optional expenses first
    options.sort(
        key=lambda option: option["current_amount"],
        reverse=True
    )

    return options
```

```python
def apply_expense_reduction(expenses, expense_name, reduction_amount, budget_period="weekly"):
    """
    Apply a confirmed reduction to an expense.

    reduction_amount is given in the current budgeting period.
    The function converts the change back to the expense's
    original frequency before updating it.
    """

    if reduction_amount < 0:
        raise ValueError("Reduction amount cannot be negative.")

    for expense in expenses:

        if expense["name"].lower() == expense_name.lower():

            if expense.get("status") != "confirmed":
                raise ValueError("Only confirmed expenses can be changed.")

            user_amount = calculate_user_share(
                expense["amount"],
                expense.get("share_percent", 100)
            )

            current_budget_amount = normalise_frequency(
                user_amount,
                expense["frequency"],
                budget_period
            )

            if reduction_amount > current_budget_amount:
                raise ValueError(
                    "Reduction cannot be greater than the current expense."
                )

            new_budget_amount = (
                current_budget_amount - reduction_amount
            )

            # Keep the original frequency correctly
            if current_budget_amount == 0:
                new_original_amount = 0
            else:
                remaining_ratio = (
                    new_budget_amount / current_budget_amount
                )

                new_original_amount = (
                    expense["amount"] * remaining_ratio
                )

            expense["amount"] = new_original_amount

            return expenses

    raise ValueError("Expense not found.")
```

```python
def recalculate_full_plan(
    general_income,
    expenses,
    goals,
    buffer_amount=0,
    budget_period="weekly"
):
    """
    Recalculate the whole financial plan using the latest
    income, expenses, goals and safety buffer.

    The original expenses and goals are not changed.
    """

    goals_copy = copy.deepcopy(goals)

    total_expenses = calculate_total_expenses(
        expenses,
        budget_period
    )

    money_remaining = calculate_money_remaining(
        general_income,
        total_expenses
    )

    # If current expenses already exceed income,
    # there is no money available for saving goals.
    if money_remaining < 0:
        return {
            "status": "budget deficit",
            "total_expenses": total_expenses,
            "money_remaining": money_remaining,
            "money_for_goals": 0,
            "total_required": 0,
            "surplus": 0,
            "shortfall": abs(money_remaining),
            "goals": goals_copy
        }

    money_for_goals = calculate_money_for_goals(
        money_remaining,
        buffer_amount
    )

    goals_copy = update_goal_requirements(
        goals_copy
    )

    goal_result = check_updated_plan(
        goals_copy,
        money_for_goals
    )

    return {
        "status": goal_result["status"],
        "total_expenses": total_expenses,
        "money_remaining": money_remaining,
        "money_for_goals": money_for_goals,
        "total_required": goal_result["total_required"],
        "surplus": goal_result["surplus"],
        "shortfall": goal_result["shortfall"],
        "goals": goals_copy
    }
```

```python
import time
import requests

def ask_gemini(prompt):
    """
    Send a prompt to Gemini.
    Retry temporary API errors instead of crashing.
    """

    headers = {
        "x-goog-api-key": GEMINI_API_KEY,
        "Content-Type": "application/json"
    }

    data = {
        "contents": [
            {
                "parts": [
                    {"text": prompt}
                ]
            }
        ]
    }

    for attempt in range(4):

        try:
            response = requests.post(
                GEMINI_URL,
                headers=headers,
                json=data,
                timeout=60
            )

            if response.status_code == 200:
                result = response.json()

                return (
                    result["candidates"][0]
                    ["content"]["parts"][0]["text"]
                )

            elif response.status_code in [408, 429, 500, 502, 503, 504]:

                wait_time = 2 ** attempt

                print(
                    f"Gemini temporarily unavailable. "
                    f"Retrying in {wait_time} seconds..."
                )

                time.sleep(wait_time)

            else:
                response.raise_for_status()

        except requests.exceptions.Timeout:

            wait_time = 2 ** attempt

            print(
                f"Gemini timed out. "
                f"Retrying in {wait_time} seconds..."
            )

            time.sleep(wait_time)

        except requests.exceptions.RequestException as error:
            return f"Gemini request failed: {error}"

    return "Gemini is temporarily unavailable. Please try again later."
```

```python
import json
import time
import requests

def extract_expenses_with_gemini(user_text):
    """
    Extract expense information from natural language using Gemini.
    Handles temporary API errors and timeouts safely.
    """

    prompt = f"""
You are helping a personal budgeting application.

Extract all expenses clearly mentioned in the user's text.

Rules:
- Do not invent missing amounts.
- Do not invent missing frequencies.
- Do not invent sharing percentages.
- If information is unknown, use null.
- Keep the expense name simple.
- Frequency must be:
  weekly, fortnightly, monthly, yearly, irregular, or null.
- Classification must be:
  fixed, essential flexible, optional flexible, or null.
- Pattern must be:
  regular, variable, seasonal, temporary, one-off, irregular, or null.

Return JSON only.

User text:
{user_text}
"""

    headers = {
        "x-goog-api-key": GEMINI_API_KEY,
        "Content-Type": "application/json"
    }

    data = {
        "contents": [
            {
                "parts": [
                    {"text": prompt}
                ]
            }
        ],
        "generationConfig": {
            "responseMimeType": "application/json"
        }
    }

    for attempt in range(4):

        try:
            response = requests.post(
                GEMINI_URL,
                headers=headers,
                json=data,
                timeout=60
            )

            if response.status_code == 200:
                result = response.json()

                response_text = (
                    result["candidates"][0]
                    ["content"]["parts"][0]["text"]
                )

                return json.loads(response_text)

            elif response.status_code in [408, 429, 500, 502, 503, 504]:

                wait_time = 2 ** attempt

                print(
                    f"Gemini is temporarily unavailable. "
                    f"Retrying in {wait_time} seconds..."
                )

                time.sleep(wait_time)

            else:
                response.raise_for_status()

        except requests.exceptions.Timeout:

            wait_time = 2 ** attempt

            print(
                f"Gemini timed out. "
                f"Retrying in {wait_time} seconds..."
            )

            time.sleep(wait_time)

        except requests.exceptions.RequestException as error:
            print("Gemini request failed:", error)
            return None

    print(
        "Gemini is currently unavailable. "
        "Please try again later."
    )

    return None
```

```python
def normalise_expense_name(name):
    """
    Convert similar expense names into one common name
    for missing-expense checking.
    """

    name = name.strip().lower()

    aliases = {
        "public transport": "transport",
        "bus": "transport",
        "train": "transport",
        "transportation": "transport",

        "mobile": "phone",
        "mobile phone": "phone",
        "phone bill": "phone",

        "food": "groceries",
        "grocery": "groceries",

        "eating out": "going out",
        "entertainment": "going out"
    }

    return aliases.get(name, name)


def find_missing_expenses(extracted_expenses, suggested_expenses):
    """
    Return suggested expenses that have not already
    been mentioned by the user.
    """

    existing_names = []

    for expense in extracted_expenses:
        existing_names.append(
            normalise_expense_name(expense["name"])
        )

    missing_expenses = []

    for expense_name in suggested_expenses:

        normalised_name = normalise_expense_name(
            expense_name
        )

        if normalised_name not in existing_names:
            missing_expenses.append(expense_name)

    return missing_expenses
```

```python
def build_expense_checklist(profile):
    """
    Build suggested expenses from the user's profile.
    Suggestions are cumulative.
    """

    suggestions = []

    def add_expense(name):
        if name not in suggestions:
            suggestions.append(name)

    # Common
    add_expense("Groceries")
    add_expense("Phone")
    add_expense("Transport")

    # Housing
    if profile.get("has_housing_cost", False):

        bills_status = profile.get("housing_bills", "unknown")

        if bills_status == "not_included":
            for bill in ["Electricity", "Water", "Gas", "Internet"]:
                add_expense(bill)

        elif bills_status == "partly_included":

            included_bills = profile.get("included_bills", [])

            for bill in ["Electricity", "Water", "Gas", "Internet"]:
                if bill not in included_bills:
                    add_expense(bill)

        elif bills_status == "unknown":
            add_expense("Housing bills")

    # Student
    if "student" in profile.get("roles", []):
        add_expense("Tuition")
        add_expense("Study materials")
        add_expense("Software")

    # Car
    if profile.get("has_car", False):
        add_expense("Fuel")
        add_expense("Car insurance")
        add_expense("Registration")
        add_expense("Servicing")

    # Children - cumulative by age group
    for child in profile.get("children", []):

        age_group = child.get("age_group")

        if age_group == "baby":
            add_expense("Nappies")
            add_expense("Baby food or formula")
            add_expense("Baby medical")
            add_expense("Childcare")

        elif age_group == "preschool":
            add_expense("Childcare")
            add_expense("Child clothing")
            add_expense("Child medical")

        elif age_group == "primary school":
            add_expense("School costs")
            add_expense("School activities")
            add_expense("Child clothing")
            add_expense("Child medical")

        elif age_group == "teen":
            add_expense("School costs")
            add_expense("Activities")
            add_expense("Child clothing")
            add_expense("Child transport")

        elif age_group == "university age":
            add_expense("Education support")
            add_expense("Transport support")

        elif age_group == "adult":
            if child.get("financially_supported", False):
                add_expense("Adult child support")

    # Pets
    for pet in profile.get("pets", []):

        add_expense("Pet food")
        add_expense("Vet")

        if pet.get("has_insurance", False):
            add_expense("Pet insurance")

        if pet.get("regular_medication", False):
            add_expense("Pet medication")

    # Other financially supported people
    for person in profile.get("dependants", []):

        responsibilities = person.get("responsibilities", [])

        if "financial support" in responsibilities:
            add_expense("Support payments")

        if "care responsibility" in responsibilities:
            add_expense("Care costs")

        if "medical support" in responsibilities:
            add_expense("Medical support")

    # Hobbies / interests
    if profile.get("has_hobbies", False):
        add_expense("Hobbies")

    return suggestions
```

```python
def extract_income_with_gemini(user_text):
    """
    Extract income sources from natural language.
    """

    prompt = f"""
You are helping a personal budgeting application.

Extract only income clearly mentioned by the user.

Important rules:

1. "amount" means the TOTAL income received from that source.
2. If the user says only part of an income is available to the budget:
   - keep the full income in "amount"
   - use availability_type = "fixed" or "percent"
   - put the usable amount or percentage in availability_value

Example:
"My partner earns $900 weekly but contributes $350 to the household."
means:
amount = 900
availability_type = "fixed"
availability_value = 350
restricted = false

3. restricted = true ONLY when the income must be used for a specific purpose,
for example:
- childcare payment only for childcare
- education grant only for study costs

A partner contribution is NOT restricted income.

4. Do not guess missing amounts or frequencies.

5. frequency must be:
weekly, fortnightly, monthly, yearly, irregular, or null.

6. availability_type must be:
all, fixed, percent, none, or null.

Return JSON only in this structure:

{{
  "income": [
    {{
      "name": "income source",
      "amount": 0,
      "frequency": "weekly",
      "availability_type": "all",
      "availability_value": null,
      "restricted": false
    }}
  ]
}}

User text:
{user_text}
"""

    data = {
        "contents": [
            {
                "parts": [
                    {"text": prompt}
                ]
            }
        ],
        "generationConfig": {
            "responseMimeType": "application/json"
        }
    }

    for attempt in range(4):

        try:
            response = requests.post(
                GEMINI_URL,
                headers={
                    "x-goog-api-key": GEMINI_API_KEY,
                    "Content-Type": "application/json"
                },
                json=data,
                timeout=60
            )

            if response.status_code == 200:
                result = response.json()

                text = (
                    result["candidates"][0]
                    ["content"]["parts"][0]["text"]
                )

                return json.loads(text)["income"]

            if response.status_code in [408, 429, 500, 502, 503, 504]:
                wait_time = 2 ** attempt
                print(f"Retrying in {wait_time} seconds...")
                time.sleep(wait_time)
                continue

            response.raise_for_status()

        except requests.exceptions.Timeout:
            wait_time = 2 ** attempt
            print(f"Gemini timed out. Retrying in {wait_time} seconds...")
            time.sleep(wait_time)

    return None
```

```python
def find_goal_followups(goals):
    """
    Find missing or unclear information before
    calculating a saving plan.
    """

    followups = []

    month_names = [
        "january", "february", "march", "april",
        "may", "june", "july", "august",
        "september", "october", "november", "december"
    ]

    for goal in goals:

        missing = []

        if goal.get("target_amount") is None:
            missing.append("target amount")

        if goal.get("current_saved") is None:
            missing.append("current savings")

        if goal.get("priority") is None:
            missing.append("priority")

        deadline = goal.get("deadline_text")

        if deadline is None:
            missing.append("deadline")

        elif deadline.strip().lower() in month_names:
            missing.append("exact deadline date")

        if missing:
            followups.append({
                "name": goal["name"],
                "missing": missing
            })

    return followups
```

```python
def use_extra_money_for_priority(
    goals,
    allocations,
    extra_money
):
    """
    Add extra money to active goals by priority
    without saving more than each goal still needs.
    """

    updated_allocations = allocations.copy()

    priority_order = {
        "must have": 1,
        "important": 2,
        "nice to have": 3
    }

    active_goals = [
        goal for goal in goals
        if goal["status"] == "active"
    ]

    active_goals.sort(
        key=lambda goal: priority_order.get(
            goal["priority"].lower(),
            99
        )
    )

    money_left = extra_money

    for goal in active_goals:

        if money_left <= 0:
            break

        current_allocation = updated_allocations.get(
            goal["name"],
            0
        )

        remaining_after_base = max(
            goal["remaining_amount"] - current_allocation,
            0
        )

        extra_for_goal = min(
            money_left,
            remaining_after_base
        )

        updated_allocations[goal["name"]] = (
            current_allocation + extra_for_goal
        )

        money_left -= extra_for_goal

    return updated_allocations, money_left
```

```python
from datetime import datetime, timedelta, date

def filter_transactions_by_week(
    transactions,
    reference_date=None
):
    """
    Return transactions from the Monday-Sunday week
    containing the reference date.
    """

    if reference_date is None:
        reference_date = date.today()

    if isinstance(reference_date, str):
        reference_date = datetime.strptime(
            reference_date,
            "%Y-%m-%d"
        ).date()

    week_start = reference_date - timedelta(
        days=reference_date.weekday()
    )

    week_end = week_start + timedelta(days=6)

    weekly_transactions = []

    for transaction in transactions:

        transaction_date = datetime.strptime(
            transaction["date"],
            "%Y-%m-%d"
        ).date()

        if week_start <= transaction_date <= week_end:
            weekly_transactions.append(transaction)

    return weekly_transactions
```

```python
def compare_actual_to_plan(
    transactions,
    planned_expenses,
    budget_period="weekly",
    period_complete=False
):
    """
    Compare actual spending with the planned budget.
    """

    actual_totals = {}

    for transaction in transactions:

        category = normalise_expense_name(
            transaction["category"]
        )

        actual_totals[category] = (
            actual_totals.get(category, 0)
            + transaction["amount"]
        )

    planned_totals = {}

    for expense in planned_expenses:

        if expense.get("status") != "confirmed":
            continue

        category = normalise_expense_name(
            expense["name"]
        )

        user_amount = calculate_user_share(
            expense["amount"],
            expense.get("share_percent", 100)
        )

        planned_amount = normalise_frequency(
            user_amount,
            expense["frequency"],
            budget_period
        )

        planned_totals[category] = (
            planned_totals.get(category, 0)
            + planned_amount
        )

    comparison = []

    for category, actual in actual_totals.items():

        planned = planned_totals.get(category)

        if planned is None:
            difference = None
            status = "unplanned expense"

        else:
            difference = actual - planned

            if difference > 0:
                status = "over budget"

            elif difference < 0:
                if period_complete:
                    status = "under budget"
                else:
                    status = "within budget so far"

            else:
                status = "on budget"

        comparison.append({
            "category": category.title(),
            "planned": planned,
            "actual": actual,
            "difference": difference,
            "status": status
        })

    return comparison
```

```python
def calculate_overspending_impact(
    spending_comparison,
    normal_money_for_goals,
    goals
):
    """
    Reduce this week's goal money by actual overspending
    and check whether the goal plan is still achievable.
    """

    total_overspending = 0

    for item in spending_comparison:

        if (
            item["difference"] is not None
            and item["difference"] > 0
        ):
            total_overspending += item["difference"]

    adjusted_money_for_goals = max(
        normal_money_for_goals - total_overspending,
        0
    )

    updated_plan = check_updated_plan(
        goals,
        adjusted_money_for_goals
    )

    return {
        "total_overspending": total_overspending,
        "money_for_goals_after_overspending":
            adjusted_money_for_goals,
        "status": updated_plan["status"],
        "surplus": updated_plan["surplus"],
        "shortfall": updated_plan["shortfall"]
    }
```

```python
def create_weekly_feedback(
    spending_comparison,
    impact
):
    """
    Create a simple weekly tracking summary
    from calculated financial results.
    """

    messages = []

    for item in spending_comparison:

        if item["status"] == "over budget":
            messages.append(
                f"{item['category']} was over budget "
                f"by ${item['difference']:.2f}."
            )

    if impact["total_overspending"] > 0:
        messages.append(
            f"Total overspending this week was "
            f"${impact['total_overspending']:.2f}."
        )

    if impact["status"] == "achievable":
        messages.append(
            "Your current saving goals are still achievable."
        )

    else:
        messages.append(
            f"Your saving plan is short by "
            f"${impact['shortfall']:.2f} this week."
        )

    return messages
```

```python
def generate_weekly_advice(feedback_messages):
    """
    Ask Gemini to rewrite verified Python results
    into a short, friendly finance summary.
    """

    facts = "\n".join(feedback_messages)

    prompt = f"""
You are a supportive personal budgeting assistant.

Rewrite the verified facts below into a short and clear weekly summary.

Rules:
- Do not change any numbers.
- Do not strengthen or weaken the verified conclusions.
- If the facts say "achievable", say "achievable", not "completely", "definitely", or "guaranteed".
- Do not invent new financial information.
- Do not claim something is wrong unless the facts say so.
- Keep the response concise and practical.
- Use a friendly but professional tone.

Verified facts:
{facts}
"""

    response = ask_gemini(prompt)

    # Fallback if Gemini is unavailable
    if (
        response is None
        or "temporarily unavailable" in response.lower()
        or "request failed" in response.lower()
    ):
        return "\n".join(feedback_messages)

    return response
```

```python
def load_transactions_csv(file_path):
    """
    Load transaction history from a CSV file.
    """

    df = pd.read_csv(file_path)

    required_columns = [
        "Date",
        "Category",
        "Amount",
        "Description"
    ]

    for column in required_columns:
        if column not in df.columns:
            raise ValueError(
                f"Missing required column: {column}"
            )

    df["Amount"] = pd.to_numeric(
        df["Amount"],
        errors="coerce"
    )

    df = df.dropna(subset=["Amount"])

    transactions = []

    for _, row in df.iterrows():
        transactions.append({
            "date": str(row["Date"]),
            "category": str(row["Category"]),
            "amount": float(row["Amount"]),
            "description": str(row["Description"])
        })

    return transactions
```

```python
def replan_from_state(state, reference_date=None):
    """
    Recalculate the full saving plan from the latest saved state.
    """

    state_copy = copy.deepcopy(state)

    # 1. Recalculate usable income
    general_income, restricted_income = calculate_income_totals(
        state_copy["income"],
        budget_period="weekly"
    )

    # 2. Recalculate weekly expenses
    total_expenses = calculate_total_expenses(
        state_copy["expenses"],
        budget_period="weekly"
    )

    # 3. Money remaining after expenses
    money_remaining = calculate_money_remaining(
        general_income,
        total_expenses
    )

    buffer_amount = state_copy.get("buffer", 0)

    money_for_goals = calculate_money_for_goals(
        money_remaining,
        buffer_amount
    )

    # 4. Recalculate goal deadlines
    goals = state_copy["goals"]

    for goal in goals:

        if (
            goal.get("status") == "active"
            and goal.get("deadline") is not None
        ):
            goal["periods_left"] = calculate_weeks_left(
                goal["deadline"],
                reference_date
            )

    # 5. Recalculate goal requirements
    goals = update_goal_requirements(goals)

    # 6. Check feasibility
    plan_result = check_updated_plan(
        goals,
        money_for_goals
    )

    # 7. Allocate weekly goal savings
    allocation_result = allocate_goal_savings(
        goals,
        money_for_goals
    )

    return {
        "general_income": general_income,
        "restricted_income": restricted_income,
        "total_expenses": total_expenses,
        "money_remaining": money_remaining,
        "buffer": buffer_amount,
        "money_for_goals": money_for_goals,
        "goals": goals,
        "status": plan_result["status"],
        "total_required": plan_result["total_required"],
        "surplus": plan_result["surplus"],
        "shortfall": plan_result["shortfall"],
        "allocations": allocation_result["allocations"],
        "extra_money": allocation_result["extra_money"]
    }
```
